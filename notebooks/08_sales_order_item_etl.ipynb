## 1.7 Independent Analyst Review

Perform a systematic independent analyst review of the untouched raw Sales Order Item datasets.

This section is the notebook-based version of the Pre-ETL Inspection.

The review goes beyond the automated profiler and checks each Sales Order Item source dataset using independent analyst validation relevant to its business meaning and approved ETL design.

The review covers, where applicable:

- Dataset structure and grain

- Missing and blank values

- Exact duplicate rows

- Primary-key candidates and key uniqueness

- Parent order relationship and order-line consistency

- Cross-source key uniqueness and collision checks

- Whitespace in text fields

- Text consistency

- Numeric validity and statistical outlier candidates

- Quantity and amount business-rule consistency

- Date parsing, datatype, validity, and Created/Modified date consistency

- ProductCode relationship values

- Source-to-target datatype compatibility considerations

- Cross-source structural consistency

Automated profiler findings and independent analyst findings are kept conceptually separate.

Each source dataset is reviewed independently before consolidation.

No data is modified in this section.

```python
# =========================================================
# 1.7 Independent Analyst Review
# =========================================================


# =========================================================
# 1. Dataset Structure and Grain
# =========================================================

print("1. DATASET STRUCTURE AND GRAIN")
print("=" * 80)

sales_order_item_sources = {
    "Shopify": shopify_order_items_raw,
    "Amazon UAE": amazon_order_items_raw,
    "Noon": noon_order_items_raw
}


for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    print(f"Rows: {source_df.shape[0]}")
    print(f"Columns: {source_df.shape[1]}")

    print("\nWorking assumption:")
    print("One row represents one product/order line within a sales order.")


# =========================================================
# 2. Missing Values and Blank Values
# =========================================================

print("\n\n2. MISSING VALUES AND BLANK VALUES")
print("=" * 80)

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    missing_values = source_df.isna().sum()

    blank_values = pd.Series(
        {
            column: (
                source_df[column]
                .astype("string")
                .str.strip()
                .eq("")
                .sum()
                if source_df[column].dtype == "object"
                else 0
            )
            for column in source_df.columns
        }
    )

    missing_blank_review = pd.DataFrame({
        "MissingValues": missing_values,
        "BlankValues": blank_values
    })

    display(missing_blank_review)


# =========================================================
# 3. Exact Duplicate Rows
# =========================================================

print("\n\n3. EXACT DUPLICATE ROWS")
print("=" * 80)

for source_name, source_df in sales_order_item_sources.items():

    exact_duplicate_count = source_df.duplicated().sum()

    print(
        f"{source_name} - Exact duplicate rows: "
        f"{exact_duplicate_count}"
    )

    if exact_duplicate_count > 0:

        display(
            source_df[
                source_df.duplicated(keep=False)
            ]
        )


# =========================================================
# 4. Primary Key Candidate and Identifier Uniqueness
# =========================================================

print("\n\n4. PRIMARY KEY CANDIDATE AND IDENTIFIER UNIQUENESS")
print("=" * 80)

identifier_columns = [
    "SourceLineID",
    "SourceOrderID",
    "SourceOrderIDInternal",
    "ProductCode"
]

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    identifier_review = pd.DataFrame(
        {
            "Column": identifier_columns,

            "NullCount": [
                source_df[column].isna().sum()
                for column in identifier_columns
            ],

            "BlankCount": [
                source_df[column]
                .astype("string")
                .str.strip()
                .eq("")
                .sum()
                for column in identifier_columns
            ],

            "DuplicateCount": [
                source_df[column].duplicated().sum()
                for column in identifier_columns
            ],

            "UniqueValues": [
                source_df[column].nunique()
                for column in identifier_columns
            ]
        }
    )

    display(identifier_review)


# =========================================================
# 5. Cross-Source Identifier Collision Review
# =========================================================

print("\n\n5. CROSS-SOURCE IDENTIFIER COLLISION REVIEW")
print("=" * 80)

cross_source_line_ids = pd.concat(
    [
        shopify_order_items_raw[
            ["SourceSystem", "SourceLineID"]
        ],

        amazon_order_items_raw[
            ["SourceSystem", "SourceLineID"]
        ],

        noon_order_items_raw[
            ["SourceSystem", "SourceLineID"]
        ]
    ],
    ignore_index=True
)


print("Total consolidated Sales Order Item records:")
print(len(cross_source_line_ids))

print("\nUnique SourceLineID values:")
print(cross_source_line_ids["SourceLineID"].nunique())


cross_source_line_duplicates = (
    cross_source_line_ids[
        cross_source_line_ids["SourceLineID"].duplicated(
            keep=False
        )
    ]
    .sort_values("SourceLineID")
)


print(
    "\nCross-source SourceLineID collisions: "
    f"{len(cross_source_line_duplicates)}"
)

if not cross_source_line_duplicates.empty:

    display(cross_source_line_duplicates)


# =========================================================
# 6. Parent Order / Order-Line Relationship Review
# =========================================================

print("\n\n6. PARENT ORDER / ORDER-LINE RELATIONSHIP REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    order_line_review = pd.DataFrame({
        "SourceOrderID": source_df["SourceOrderID"],
        "SourceOrderIDInternal": source_df["SourceOrderIDInternal"],
        "SourceLineID": source_df["SourceLineID"]
    })

    print("Distinct parent orders:")
    print(source_df["SourceOrderID"].nunique())

    print("Distinct order-line IDs:")
    print(source_df["SourceLineID"].nunique())

    print("Orders containing multiple line items:")

    multi_line_orders = (
        source_df
        .groupby("SourceOrderID")
        .size()
        .reset_index(name="LineItemCount")
    )

    multi_line_orders = multi_line_orders[
        multi_line_orders["LineItemCount"] > 1
    ]

    display(multi_line_orders)


# =========================================================
# 7. Leading and Trailing Whitespace Review
# =========================================================

print("\n\n7. LEADING AND TRAILING WHITESPACE REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    text_columns = source_df.select_dtypes(
        include="object"
    ).columns

    whitespace_findings = []

    for column in text_columns:

        whitespace_mask = (
            source_df[column].astype("string")
            != source_df[column].astype("string").str.strip()
        )

        count = whitespace_mask.sum()

        if count > 0:

            whitespace_records = source_df.loc[
                whitespace_mask,
                ["SourceLineID", column]
            ]

            whitespace_findings.append({
                "Column": column,
                "WhitespaceRecordCount": count
            })

            print(f"\nColumn: {column}")
            display(whitespace_records)

    if whitespace_findings:

        print("\nWhitespace findings summary:")
        display(pd.DataFrame(whitespace_findings))

    else:

        print("No leading or trailing whitespace found.")


# =========================================================
# 8. Text Consistency Review
# =========================================================

print("\n\n8. TEXT CONSISTENCY REVIEW")
print("=" * 80)

text_consistency_columns = [
    "SourceSystem",
    "ProductCode",
    "CreatedByUser",
    "ModifiedByUser"
]

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    for column in text_consistency_columns:

        print(f"\n{column} values:")

        display(
            source_df[column]
            .value_counts(dropna=False)
            .sort_index()
            .rename("RecordCount")
            .to_frame()
        )


# =========================================================
# 9. Numeric Validity Review
# =========================================================

print("\n\n9. NUMERIC VALIDITY REVIEW")
print("=" * 80)

numeric_columns = [
    "Qty",
    "UnitPrice",
    "DiscountAmount",
    "LineAmount"
]

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    numeric_summary = source_df[
        numeric_columns
    ].describe().T

    numeric_summary["NegativeCount"] = [
        (source_df[column] < 0).sum()
        for column in numeric_columns
    ]

    numeric_summary["ZeroCount"] = [
        (source_df[column] == 0).sum()
        for column in numeric_columns
    ]

    numeric_summary["PositiveCount"] = [
        (source_df[column] > 0).sum()
        for column in numeric_columns
    ]

    display(numeric_summary)


# =========================================================
# 10. Numeric Outlier Candidate Review
# =========================================================

print("\n\n10. NUMERIC OUTLIER CANDIDATE REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    outlier_summary = []

    for column in numeric_columns:

        q1 = source_df[column].quantile(0.25)
        q3 = source_df[column].quantile(0.75)

        iqr = q3 - q1

        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)

        outlier_mask = (
            (source_df[column] < lower_bound)
            |
            (source_df[column] > upper_bound)
        )

        outlier_count = outlier_mask.sum()

        outlier_summary.append({
            "Column": column,
            "LowerBoundary": lower_bound,
            "UpperBoundary": upper_bound,
            "OutlierCandidateCount": outlier_count
        })

        if outlier_count > 0:

            print(f"\nOutlier candidate(s): {column}")

            display(
                source_df.loc[
                    outlier_mask,
                    [
                        "SourceOrderID",
                        "SourceOrderIDInternal",
                        "SourceLineID",
                        "ProductCode",
                        column
                    ]
                ]
            )

    print("\nOutlier candidate summary:")
    display(pd.DataFrame(outlier_summary))


# =========================================================
# 11. Quantity and Amount Business Rule Review
# =========================================================

print("\n\n11. QUANTITY AND AMOUNT BUSINESS RULE REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    invalid_quantity = source_df[
        source_df["Qty"] <= 0
    ]

    invalid_discount = source_df[
        source_df["DiscountAmount"] > (
            source_df["Qty"] * source_df["UnitPrice"]
        )
    ]

    calculated_line_amount = (
        source_df["Qty"] * source_df["UnitPrice"]
        - source_df["DiscountAmount"]
    )

    amount_mismatch = source_df[
        calculated_line_amount != source_df["LineAmount"]
    ]

    print(
        "Records where Qty <= 0:",
        len(invalid_quantity)
    )

    print(
        "Records where DiscountAmount exceeds gross line value:",
        len(invalid_discount)
    )

    print(
        "Records where LineAmount does not equal "
        "(Qty × UnitPrice) - DiscountAmount:",
        len(amount_mismatch)
    )

    if not invalid_quantity.empty:

        print("\nInvalid quantity records:")
        display(invalid_quantity)

    if not invalid_discount.empty:

        print("\nInvalid discount records:")
        display(invalid_discount)

    if not amount_mismatch.empty:

        print("\nLine amount mismatch records:")

        mismatch_review = source_df.loc[
            amount_mismatch.index,
            [
                "SourceOrderID",
                "SourceLineID",
                "ProductCode",
                "Qty",
                "UnitPrice",
                "DiscountAmount",
                "LineAmount"
            ]
        ].copy()

        mismatch_review["CalculatedLineAmount"] = (
            calculated_line_amount.loc[
                amount_mismatch.index
            ]
        )

        display(mismatch_review)


# =========================================================
# 12. Date Parsing and Datatype Review
# =========================================================

print("\n\n12. DATE PARSING AND DATATYPE REVIEW")
print("=" * 80)

date_columns = [
    "CreatedOn",
    "ModifiedOn"
]

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    date_review = []

    for column in date_columns:

        parsed_dates = pd.to_datetime(
            source_df[column],
            errors="coerce"
        )

        date_review.append({
            "Column": column,
            "CurrentSourceDtype": str(
                source_df[column].dtype
            ),
            "InvalidDateCount": parsed_dates.isna().sum(),
            "MinimumDate": parsed_dates.min(),
            "MaximumDate": parsed_dates.max()
        })

    display(pd.DataFrame(date_review))


# =========================================================
# 13. CreatedOn / ModifiedOn Date Logic
# =========================================================

print("\n\n13. CREATEDON / MODIFIEDON DATE LOGIC REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    created_dates = pd.to_datetime(
        source_df["CreatedOn"],
        errors="coerce"
    )

    modified_dates = pd.to_datetime(
        source_df["ModifiedOn"],
        errors="coerce"
    )

    modified_before_created = source_df[
        modified_dates < created_dates
    ]

    print(
        "Records where ModifiedOn is earlier than CreatedOn: "
        f"{len(modified_before_created)}"
    )

    if not modified_before_created.empty:

        print("\nModifiedOn earlier than CreatedOn:")

        display(
            modified_before_created[
                [
                    "SourceOrderID",
                    "SourceLineID",
                    "CreatedOn",
                    "ModifiedOn"
                ]
            ]
        )


# =========================================================
# 14. ProductCode Relationship Value Review
# =========================================================

print("\n\n14. PRODUCTCODE RELATIONSHIP VALUE REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    print("\nProductCode frequency:")

    display(
        source_df["ProductCode"]
        .value_counts(dropna=False)
        .sort_index()
        .rename("RecordCount")
        .to_frame()
    )


# =========================================================
# 15. Source Datatype Review
# =========================================================

print("\n\n15. SOURCE DATATYPE REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_item_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    datatype_review = pd.DataFrame({
        "Column": source_df.columns,
        "SourceDtype": [
            str(source_df[column].dtype)
            for column in source_df.columns
        ]
    })

    display(datatype_review)


# =========================================================
# 16. Cross-Source Structural Consistency Review
# =========================================================

print("\n\n16. CROSS-SOURCE STRUCTURAL CONSISTENCY REVIEW")
print("=" * 80)

source_structure_review = pd.DataFrame(
    {
        "Source": list(sales_order_item_sources.keys()),

        "Rows": [
            source_df.shape[0]
            for source_df in sales_order_item_sources.values()
        ],

        "Columns": [
            source_df.shape[1]
            for source_df in sales_order_item_sources.values()
        ],

        "ColumnOrderMatchesShopify": [
            list(source_df.columns)
            == list(shopify_order_items_raw.columns)
            for source_df in sales_order_item_sources.values()
        ]
    }
)

display(source_structure_review)

print("\nExpected Sales Order Item source columns:")
print(shopify_order_items_raw.columns.tolist())


# =========================================================
# Independent Analyst Review Complete
# =========================================================

print("\n" + "=" * 80)
print("Independent Analyst Review completed.")
print("No data was modified.")
print("=" * 80)
```

## 1.8 Findings and Transformation Decisions

Compare the automated profiler findings, the Independent Analyst Review findings, and the approved Sales Order Item data requirements to formally determine the required treatment for each finding.

Each finding must be classified as one of the following:

- Safe deterministic transformation

- Valid value — retain unchanged

- Expected business behavior

- Requires relationship validation later

- Requires business investigation

No transformation is performed in this section.

For Sales Order Item, the approved business rules include:

- Each row represents one product/order line within a sales order.
- One sales order may contain multiple sales order items.
- `SourceLineID` represents the source line-level identifier and must remain unique.
- Quantity must be greater than zero.
- Unit price, discount amount, and line amount must be preserved as transaction values.
- `LineAmount` must equal `(Qty × UnitPrice) - DiscountAmount`.
- Historical unit price values must be preserved.
- Product relationships will be validated against the Product master during the appropriate validation stage.

## 2.2 Apply Approved Transformations

Apply only the transformations approved in Section 1.8 to the clean working Sales Order Item datasets.

Approved transformations:

- Convert CreatedOn to datetime.

- Convert ModifiedOn to datetime.

- Convert UnitPrice to Decimal-compatible values.

- Convert DiscountAmount to Decimal-compatible values.

- Convert LineAmount to Decimal-compatible values.

The datetime and monetary datatype transformations are deterministic and prepare the source values for the approved Sales Order Item target structure and downstream database loading.

This validation checks:

- CreatedOn datetime conversion

- ModifiedOn datetime conversion

- UnitPrice Decimal-compatible conversion

- DiscountAmount Decimal-compatible conversion

- LineAmount Decimal-compatible conversion

- Quantity and monetary business values remain unchanged

- LineAmount calculation remains valid after transformation

Only approved transformations and approved source-value retention decisions are validated.

No additional data modifications are performed in this section.

```python
# =========================================================
# 3.1 Transformation Validation
# =========================================================

validation_results = []


# ---------------------------------------------------------
# 1. CreatedOn Datetime Validation
# ---------------------------------------------------------

for source_name, df in [
    ("Shopify", shopify_order_items_clean),
    ("Amazon UAE", amazon_order_items_clean),
    ("Noon", noon_order_items_clean)
]:

    created_on_is_datetime = pd.api.types.is_datetime64_any_dtype(
        df["CreatedOn"]
    )

    validation_results.append({
        "Validation": f"{source_name} CreatedOn datetime conversion",
        "Expected": "datetime64 datatype",
        "Actual": str(df["CreatedOn"].dtype),
        "Passed": created_on_is_datetime
    })


# ---------------------------------------------------------
# 2. ModifiedOn Datetime Validation
# ---------------------------------------------------------

for source_name, df in [
    ("Shopify", shopify_order_items_clean),
    ("Amazon UAE", amazon_order_items_clean),
    ("Noon", noon_order_items_clean)
]:

    modified_on_is_datetime = pd.api.types.is_datetime64_any_dtype(
        df["ModifiedOn"]
    )

    validation_results.append({
        "Validation": f"{source_name} ModifiedOn datetime conversion",
        "Expected": "datetime64 datatype",
        "Actual": str(df["ModifiedOn"].dtype),
        "Passed": modified_on_is_datetime
    })


# ---------------------------------------------------------
# 3. UnitPrice Decimal Validation
# ---------------------------------------------------------

for source_name, df in [
    ("Shopify", shopify_order_items_clean),
    ("Amazon UAE", amazon_order_items_clean),
    ("Noon", noon_order_items_clean)
]:

    unit_price_is_decimal = df["UnitPrice"].apply(
        lambda value: isinstance(value, Decimal)
    ).all()

    validation_results.append({
        "Validation": f"{source_name} UnitPrice Decimal conversion",
        "Expected": "All values are Decimal-compatible",
        "Actual": (
            "All values are Decimal"
            if unit_price_is_decimal
            else "One or more values are not Decimal"
        ),
        "Passed": unit_price_is_decimal
    })


# ---------------------------------------------------------
# 4. DiscountAmount Decimal Validation
# ---------------------------------------------------------

for source_name, df in [
    ("Shopify", shopify_order_items_clean),
    ("Amazon UAE", amazon_order_items_clean),
    ("Noon", noon_order_items_clean)
]:

    discount_amount_is_decimal = df["DiscountAmount"].apply(
        lambda value: isinstance(value, Decimal)
    ).all()

    validation_results.append({
        "Validation": f"{source_name} DiscountAmount Decimal conversion",
        "Expected": "All values are Decimal-compatible",
        "Actual": (
            "All values are Decimal"
            if discount_amount_is_decimal
            else "One or more values are not Decimal"
        ),
        "Passed": discount_amount_is_decimal
    })


# ---------------------------------------------------------
# 5. LineAmount Decimal Validation
# ---------------------------------------------------------

for source_name, df in [
    ("Shopify", shopify_order_items_clean),
    ("Amazon UAE", amazon_order_items_clean),
    ("Noon", noon_order_items_clean)
]:

    line_amount_is_decimal = df["LineAmount"].apply(
        lambda value: isinstance(value, Decimal)
    ).all()

    validation_results.append({
        "Validation": f"{source_name} LineAmount Decimal conversion",
        "Expected": "All values are Decimal-compatible",
        "Actual": (
            "All values are Decimal"
            if line_amount_is_decimal
            else "One or more values are not Decimal"
        ),
        "Passed": line_amount_is_decimal
    })


# ---------------------------------------------------------
# 6. Quantity and Monetary Business Value Validation
# ---------------------------------------------------------

for source_name, df in [
    ("Shopify", shopify_order_items_clean),
    ("Amazon UAE", amazon_order_items_clean),
    ("Noon", noon_order_items_clean)
]:

    invalid_quantity_count = (
        df["Qty"] <= 0
    ).sum()

    invalid_discount_count = (
        df["DiscountAmount"]
        > (
            df["Qty"].apply(Decimal)
            * df["UnitPrice"]
        )
    ).sum()

    validation_results.append({
        "Validation": f"{source_name} quantity and monetary business values",
        "Expected": (
            "Qty > 0 and DiscountAmount does not exceed "
            "gross line value"
        ),
        "Actual": (
            f"Qty <= 0: {invalid_quantity_count}; "
            f"Discount exceeds gross line value: "
            f"{invalid_discount_count}"
        ),
        "Passed": (
            invalid_quantity_count == 0
            and invalid_discount_count == 0
        )
    })


# ---------------------------------------------------------
# 7. LineAmount Calculation Validation
# ---------------------------------------------------------

for source_name, df in [
    ("Shopify", shopify_order_items_clean),
    ("Amazon UAE", amazon_order_items_clean),
    ("Noon", noon_order_items_clean)
]:

    calculated_line_amount = (
        df["Qty"].apply(Decimal)
        * df["UnitPrice"]
        - df["DiscountAmount"]
    )

    line_amount_mismatch_count = (
        calculated_line_amount
        != df["LineAmount"]
    ).sum()

    validation_results.append({
        "Validation": f"{source_name} LineAmount calculation",
        "Expected": (
            "LineAmount equals "
            "(Qty × UnitPrice) - DiscountAmount"
        ),
        "Actual": (
            f"{line_amount_mismatch_count} calculation mismatches"
        ),
        "Passed": line_amount_mismatch_count == 0
    })


# ---------------------------------------------------------
# 8. Profiler Outlier Candidate Retention Validation
# ---------------------------------------------------------

for source_name, raw_df, clean_df in [
    ("Shopify", shopify_order_items_raw, shopify_order_items_clean),
    ("Amazon UAE", amazon_order_items_raw, amazon_order_items_clean),
    ("Noon", noon_order_items_raw, noon_order_items_clean)
]:

    source_line_ids = raw_df["SourceLineID"].tolist()

    retained_line_ids = clean_df["SourceLineID"].tolist()

    line_ids_retained = (
        source_line_ids == retained_line_ids
    )

    validation_results.append({
        "Validation": (
            f"{source_name} profiler outlier candidate "
            "retention"
        ),
        "Expected": (
            "Source records and identified outlier candidates "
            "remain retained"
        ),
        "Actual": (
            "Source line records retained unchanged"
            if line_ids_retained
            else "Source line records changed or removed"
        ),
        "Passed": line_ids_retained
    })


# ---------------------------------------------------------
# Transformation Validation Results
# ---------------------------------------------------------

transformation_validation = pd.DataFrame(
    validation_results
)

print("TRANSFORMATION VALIDATION RESULTS")
print("=" * 120)

display(transformation_validation)


# ---------------------------------------------------------
# Overall Result
# ---------------------------------------------------------

overall_validation_passed = (
    transformation_validation["Passed"].all()
)

print(
    "\nOverall Transformation Validation Passed:",
    overall_validation_passed
)

if not overall_validation_passed:
    raise ValueError(
        "Transformation validation failed. Review the results "
        "before proceeding."
    )
```

## 6.2.1 Create Sales Order Item Table

The `commerce.sales_order_item` table does not currently exist.

This section creates the Sales Order Item PostgreSQL target table based

on the approved LJ Dev Commerce Sales Order Item Data Dictionary.

Key design decisions:

- `sales_order_item_id` is the Primary Key.

- All required Data Dictionary fields use `NOT NULL`.

- `sales_order_id` and `product_id` are Foreign Keys based on the

  approved Sales Order Item database relationships.

- `created_date` and `updated_date` use the PostgreSQL

  `TIMESTAMP` datatype.

- `unit_price`, `discount_amount`, and `line_total` use the PostgreSQL

  `DECIMAL` datatype.

```python
# =========================================================
# 6.2.1 Create Sales Order Item Table
# =========================================================

create_sales_order_item_table_query = """
CREATE TABLE commerce.sales_order_item (

    sales_order_item_id TEXT PRIMARY KEY,

    sales_order_id TEXT NOT NULL,
    product_id TEXT NOT NULL,

    quantity INTEGER NOT NULL,

    unit_price DECIMAL NOT NULL,
    discount_amount DECIMAL NOT NULL,
    line_total DECIMAL NOT NULL,

    created_date TIMESTAMP NOT NULL,
    created_by TEXT NOT NULL,

    updated_date TIMESTAMP NOT NULL,
    updated_by TEXT NOT NULL,

    FOREIGN KEY (sales_order_id)
        REFERENCES commerce.sales_order(sales_order_id),

    FOREIGN KEY (product_id)
        REFERENCES commerce.product(product_id)

);
"""

try:
    cursor.execute(create_sales_order_item_table_query)
    conn.commit()

    print("Sales Order Item table created successfully.")
    print("Table: commerce.sales_order_item")

except Exception as e:
    conn.rollback()

    print("Sales Order Item table creation failed.")
    print("Error:", e)
```

- Commit the transaction only after successful insertion.

No transformations are performed during the database loading step.

```python
# =========================================================
# 6.5 Insert Sales Order Item Records
# =========================================================

print("Sales Order Item Record Loading")
print("=" * 80)


# ---------------------------------------------------------
# 1. Confirm Target Table Is Empty
# ---------------------------------------------------------

cursor.execute("""
    SELECT COUNT(*)
    FROM commerce.sales_order_item;
""")

existing_record_count = cursor.fetchone()[0]

print("\n1. Target Table Record Count")
print("Existing records:", existing_record_count)

target_table_empty = (
    existing_record_count == 0
)

print("Target table is empty:", target_table_empty)


# ---------------------------------------------------------
# 2. Confirm Database-Ready Row Count
# ---------------------------------------------------------

db_ready_row_count = len(
    sales_order_item_db_ready
)

print("\n2. Database-Ready Row Count")
print("Database-ready records:", db_ready_row_count)


# ---------------------------------------------------------
# 3. Insert Validated Database-Ready Records
# ---------------------------------------------------------

insert_sales_order_item_query = """
INSERT INTO commerce.sales_order_item (
    sales_order_id,
    sales_order_item_id,
    product_id,
    quantity,
    unit_price,
    discount_amount,
    line_total,
    created_date,
    created_by,
    updated_date,
    updated_by
)
VALUES (
    %s, %s, %s, %s, %s, %s, %s,
    %s, %s, %s, %s
);
"""

try:

    if not target_table_empty:
        raise ValueError(
            "Target table is not empty. "
            "Insertion stopped to prevent accidental duplicate loading."
        )

    records_to_insert = [
        tuple(row)
        for row in sales_order_item_db_ready.itertuples(
            index=False,
            name=None
        )
    ]

    if len(records_to_insert) != db_ready_row_count:
        raise ValueError(
            "Database-ready record count changed before insertion."
        )

    cursor.executemany(
        insert_sales_order_item_query,
        records_to_insert
    )

    conn.commit()

    print("\n3. Insert Result")
    print(
        "Sales Order Item records inserted:",
        len(records_to_insert)
    )
    print("Transaction committed successfully.")

except Exception as e:

    conn.rollback()

    print("\nSales Order Item record insertion failed.")
    print("Transaction rolled back.")
    print("Error:", e)
```

## 6.9 Sales Order Item Database Integrity Validation

Perform final integrity checks on the loaded

`commerce.sales_order_item` table.

The checks confirm:

- Primary-key uniqueness

- Required-field integrity

- Duplicate record checks

- Quantity validity

- Line amount integrity

- Audit date consistency

These checks provide the final confirmation that the Sales Order Item

table is structurally and business-rule compliant after loading.

```python
# =========================================================
# 6.9 Sales Order Item Database Integrity Validation
# =========================================================

integrity_query = """
SELECT
    COUNT(*) AS total_rows,

    COUNT(DISTINCT sales_order_item_id) AS distinct_sales_order_item_ids,

    COUNT(*) FILTER (
        WHERE sales_order_item_id IS NULL
    ) AS null_sales_order_item_ids,

    COUNT(*) FILTER (
        WHERE sales_order_id IS NULL
    ) AS null_sales_order_ids,

    COUNT(*) FILTER (
        WHERE product_id IS NULL
    ) AS null_product_ids,

    COUNT(*) FILTER (
        WHERE quantity IS NULL
    ) AS null_quantities,

    COUNT(*) FILTER (
        WHERE unit_price IS NULL
    ) AS null_unit_prices,

    COUNT(*) FILTER (
        WHERE discount_amount IS NULL
    ) AS null_discount_amounts,

    COUNT(*) FILTER (
        WHERE line_total IS NULL
    ) AS null_line_totals,

    COUNT(*) FILTER (
        WHERE created_date IS NULL
    ) AS null_created_dates,

    COUNT(*) FILTER (
        WHERE created_by IS NULL
    ) AS null_created_by,

    COUNT(*) FILTER (
        WHERE updated_date IS NULL
    ) AS null_updated_dates,

    COUNT(*) FILTER (
        WHERE updated_by IS NULL
    ) AS null_updated_by,

    COUNT(*) FILTER (
        WHERE quantity <= 0
    ) AS invalid_quantities,

    COUNT(*) FILTER (
        WHERE unit_price < 0
    ) AS invalid_unit_prices,

    COUNT(*) FILTER (
        WHERE discount_amount < 0
        OR discount_amount > (quantity * unit_price)
    ) AS invalid_discount_amounts,

    COUNT(*) FILTER (
        WHERE line_total != (
            quantity * unit_price
            - discount_amount
        )
    ) AS invalid_line_totals,

    COUNT(*) FILTER (
        WHERE updated_date < created_date
    ) AS invalid_audit_date_order

FROM commerce.sales_order_item;
"""

try:
    cursor.execute(integrity_query)

    integrity_result = cursor.fetchone()

    print("Sales Order Item Database Integrity Validation")
    print("=" * 80)

    print("\n1. Primary Key Integrity")
    print("Total rows:", integrity_result[0])
    print(
        "Distinct sales order item IDs:",
        integrity_result[1]
    )
    print(
        "NULL sales order item IDs:",
        integrity_result[2]
    )

    print("\n2. Required Field Integrity")
    print("NULL sales order IDs:", integrity_result[3])
    print("NULL product IDs:", integrity_result[4])
    print("NULL quantities:", integrity_result[5])
    print("NULL unit prices:", integrity_result[6])
    print("NULL discount amounts:", integrity_result[7])
    print("NULL line totals:", integrity_result[8])
    print("NULL created dates:", integrity_result[9])
    print("NULL created by:", integrity_result[10])
    print("NULL updated dates:", integrity_result[11])
    print("NULL updated by:", integrity_result[12])

    print("\n3. Sales Order Item Value Integrity")
    print("Invalid quantities:", integrity_result[13])
    print("Invalid unit prices:", integrity_result[14])
    print(
        "Invalid discount amounts:",
        integrity_result[15]
    )
    print("Invalid line totals:", integrity_result[16])

    print("\n4. Audit Date Integrity")
    print(
        "Updated date earlier than created date:",
        integrity_result[17]
    )

    # -----------------------------------------------------
    # Final Validation Result
    # -----------------------------------------------------

    integrity_passed = (
        integrity_result[0] == integrity_result[1]
        and integrity_result[2] == 0
        and integrity_result[3] == 0
        and integrity_result[4] == 0
        and integrity_result[5] == 0
        and integrity_result[6] == 0
        and integrity_result[7] == 0
        and integrity_result[8] == 0
        and integrity_result[9] == 0
        and integrity_result[10] == 0
        and integrity_result[11] == 0
        and integrity_result[12] == 0
        and integrity_result[13] == 0
        and integrity_result[14] == 0
        and integrity_result[15] == 0
        and integrity_result[16] == 0
        and integrity_result[17] == 0
    )

    print("\n5. Database Integrity Validation Summary")
    print("=" * 80)

    print(
        "Sales Order Item database integrity passed:",
        integrity_passed
    )

except Exception as e:
    print(
        "Sales Order Item database integrity validation failed."
    )
    print("Error:", e)
```
