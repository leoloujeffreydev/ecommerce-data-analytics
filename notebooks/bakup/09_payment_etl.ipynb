## 1.7 Independent Analyst Review

Perform a systematic independent analyst review of the untouched raw Payment dataset.

This section is the notebook-based version of the Pre-ETL Inspection.

The review goes beyond the automated profiler and checks the Payment dataset using independent analyst validation relevant to its business meaning and approved ETL design.

The review covers, where applicable:

- Dataset structure and grain

- Missing and blank values

- Exact duplicate rows

- Primary-key candidates and key uniqueness

- Parent order relationship and payment-to-order consistency

- Whitespace in text fields

- Text consistency

- Payment method and payment status consistency

- Numeric validity and statistical outlier candidates

- Payment amount and payment status business-rule consistency

- Date parsing, datatype, validity, and Created/Modified date consistency

- Source-to-target datatype compatibility considerations

- Payment relationship values

Automated profiler findings and independent analyst findings are kept conceptually separate.

The Payment dataset is reviewed independently before any transformation.

No data is modified in this section.

```python
# =========================================================
# 1.7 Independent Analyst Review
# =========================================================


# =========================================================
# 1. Dataset Structure and Grain
# =========================================================

print("1. DATASET STRUCTURE AND GRAIN")
print("=" * 80)

print(f"Rows: {zoho_books_customer_payments_raw.shape[0]}")
print(f"Columns: {zoho_books_customer_payments_raw.shape[1]}")

print("\nWorking assumption:")
print("One row represents one payment transaction.")


# =========================================================
# 2. Missing Values and Blank Values
# =========================================================

print("\n\n2. MISSING VALUES AND BLANK VALUES")
print("=" * 80)

source_df = zoho_books_customer_payments_raw

missing_values = source_df.isna().sum()

blank_values = pd.Series(
    {
        column: (
            source_df[column]
            .astype("string")
            .str.strip()
            .eq("")
            .sum()
            if source_df[column].dtype == "object"
            else 0
        )
        for column in source_df.columns
    }
)

missing_blank_review = pd.DataFrame({
    "MissingValues": missing_values,
    "BlankValues": blank_values
})

display(missing_blank_review)


# =========================================================
# 3. Exact Duplicate Rows
# =========================================================

print("\n\n3. EXACT DUPLICATE ROWS")
print("=" * 80)

exact_duplicate_count = source_df.duplicated().sum()

print(
    "Payment - Exact duplicate rows:",
    exact_duplicate_count
)

if exact_duplicate_count > 0:

    display(
        source_df[
            source_df.duplicated(keep=False)
        ]
    )


# =========================================================
# 4. Primary Key Candidate and Identifier Uniqueness
# =========================================================

print("\n\n4. PRIMARY KEY CANDIDATE AND IDENTIFIER UNIQUENESS")
print("=" * 80)

identifier_columns = [
    "PaymentReference",
    "SourceOrderReference",
    "GatewayTransactionID"
]

identifier_review = pd.DataFrame(
    {
        "Column": identifier_columns,

        "NullCount": [
            source_df[column].isna().sum()
            for column in identifier_columns
        ],

        "BlankCount": [
            source_df[column]
            .astype("string")
            .str.strip()
            .eq("")
            .sum()
            for column in identifier_columns
        ],

        "DuplicateCount": [
            source_df[column].duplicated().sum()
            for column in identifier_columns
        ],

        "UniqueValues": [
            source_df[column].nunique()
            for column in identifier_columns
        ]
    }
)

display(identifier_review)


# =========================================================
# 5. Payment Transaction Identifier Collision Review
# =========================================================

print("\n\n5. PAYMENT TRANSACTION IDENTIFIER COLLISION REVIEW")
print("=" * 80)

transaction_identifier_review = source_df[
    [
        "PaymentReference",
        "GatewayTransactionID"
    ]
].copy()

duplicate_payment_references = (
    transaction_identifier_review[
        transaction_identifier_review["PaymentReference"]
        .duplicated(keep=False)
    ]
)

duplicate_gateway_transactions = (
    transaction_identifier_review[
        transaction_identifier_review["GatewayTransactionID"]
        .duplicated(keep=False)
    ]
)

print(
    "Duplicate PaymentReference records:",
    len(duplicate_payment_references)
)

print(
    "Duplicate GatewayTransactionID records:",
    len(duplicate_gateway_transactions)
)

if not duplicate_payment_references.empty:

    print("\nDuplicate PaymentReference values:")
    display(duplicate_payment_references)

if not duplicate_gateway_transactions.empty:

    print("\nDuplicate GatewayTransactionID values:")
    display(duplicate_gateway_transactions)


# =========================================================
# 6. Parent Order / Payment Relationship Review
# =========================================================

print("\n\n6. PARENT ORDER / PAYMENT RELATIONSHIP REVIEW")
print("=" * 80)

payment_order_review = source_df[
    [
        "SourceOrderReference",
        "PaymentReference",
        "PaymentStatus",
        "Amount"
    ]
].copy()

print("Distinct parent orders:")
print(source_df["SourceOrderReference"].nunique())

print("Distinct payment references:")
print(source_df["PaymentReference"].nunique())

print("\nOrders containing multiple payment records:")

multiple_payment_orders = (
    source_df
    .groupby("SourceOrderReference")
    .size()
    .reset_index(name="PaymentCount")
)

multiple_payment_orders = multiple_payment_orders[
    multiple_payment_orders["PaymentCount"] > 1
]

display(multiple_payment_orders)


# =========================================================
# 7. Leading and Trailing Whitespace Review
# =========================================================

print("\n\n7. LEADING AND TRAILING WHITESPACE REVIEW")
print("=" * 80)

text_columns = source_df.select_dtypes(
    include="object"
).columns

whitespace_findings = []

for column in text_columns:

    whitespace_mask = (
        source_df[column].astype("string")
        != source_df[column].astype("string").str.strip()
    )

    count = whitespace_mask.sum()

    if count > 0:

        whitespace_records = source_df.loc[
            whitespace_mask,
            ["PaymentReference", column]
        ]

        whitespace_findings.append({
            "Column": column,
            "WhitespaceRecordCount": count
        })

        print(f"\nColumn: {column}")
        display(whitespace_records)

if whitespace_findings:

    print("\nWhitespace findings summary:")
    display(pd.DataFrame(whitespace_findings))

else:

    print("No leading or trailing whitespace found.")


# =========================================================
# 8. Text Consistency Review
# =========================================================

print("\n\n8. TEXT CONSISTENCY REVIEW")
print("=" * 80)

text_consistency_columns = [
    "PaymentMethod",
    "PaymentStatus",
    "PaymentNotes",
    "CreatedByUser",
    "ModifiedByUser"
]

for column in text_consistency_columns:

    print(f"\n{column} values:")

    display(
        source_df[column]
        .value_counts(dropna=False)
        .sort_index()
        .rename("RecordCount")
        .to_frame()
    )


# =========================================================
# 9. Numeric Validity Review
# =========================================================

print("\n\n9. NUMERIC VALIDITY REVIEW")
print("=" * 80)

numeric_columns = [
    "Amount"
]

numeric_summary = source_df[
    numeric_columns
].describe().T

numeric_summary["NegativeCount"] = [
    (source_df[column] < 0).sum()
    for column in numeric_columns
]

numeric_summary["ZeroCount"] = [
    (source_df[column] == 0).sum()
    for column in numeric_columns
]

numeric_summary["PositiveCount"] = [
    (source_df[column] > 0).sum()
    for column in numeric_columns
]

display(numeric_summary)


# =========================================================
# 10. Numeric Outlier Candidate Review
# =========================================================

print("\n\n10. NUMERIC OUTLIER CANDIDATE REVIEW")
print("=" * 80)

outlier_summary = []

for column in numeric_columns:

    q1 = source_df[column].quantile(0.25)
    q3 = source_df[column].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)

    outlier_mask = (
        (source_df[column] < lower_bound)
        |
        (source_df[column] > upper_bound)
    )

    outlier_count = outlier_mask.sum()

    outlier_summary.append({
        "Column": column,
        "LowerBoundary": lower_bound,
        "UpperBoundary": upper_bound,
        "OutlierCandidateCount": outlier_count
    })

    if outlier_count > 0:

        print(f"\nOutlier candidate(s): {column}")

        display(
            source_df.loc[
                outlier_mask,
                [
                    "PaymentReference",
                    "SourceOrderReference",
                    column
                ]
            ]
        )

print("\nOutlier candidate summary:")
display(pd.DataFrame(outlier_summary))


# =========================================================
# 11. Payment Amount and Status Business Rule Review
# =========================================================

print("\n\n11. PAYMENT AMOUNT AND STATUS BUSINESS RULE REVIEW")
print("=" * 80)

invalid_negative_amount = source_df[
    source_df["Amount"] < 0
]

successful_non_positive = source_df[
    (source_df["PaymentStatus"] == "Successful")
    &
    (source_df["Amount"] <= 0)
]

pending_non_zero = source_df[
    (source_df["PaymentStatus"] == "Pending")
    &
    (source_df["Amount"] != 0)
]

failed_non_zero = source_df[
    (source_df["PaymentStatus"] == "Failed")
    &
    (source_df["Amount"] != 0)
]

print(
    "Records where Amount < 0:",
    len(invalid_negative_amount)
)

print(
    "Successful payments where Amount <= 0:",
    len(successful_non_positive)
)

print(
    "Pending payments where Amount != 0:",
    len(pending_non_zero)
)

print(
    "Failed payments where Amount != 0:",
    len(failed_non_zero)
)

if not invalid_negative_amount.empty:

    print("\nNegative payment amounts:")
    display(invalid_negative_amount)

if not successful_non_positive.empty:

    print("\nSuccessful payments with non-positive amounts:")
    display(successful_non_positive)

if not pending_non_zero.empty:

    print("\nPending payments with non-zero amounts:")
    display(pending_non_zero)

if not failed_non_zero.empty:

    print("\nFailed payments with non-zero amounts:")
    display(failed_non_zero)


# =========================================================
# 12. Date Parsing and Datatype Review
# =========================================================

print("\n\n12. DATE PARSING AND DATATYPE REVIEW")
print("=" * 80)

date_columns = [
    "PaymentDate",
    "CreatedOn",
    "ModifiedOn"
]

date_review = []

for column in date_columns:

    parsed_dates = pd.to_datetime(
        source_df[column],
        errors="coerce"
    )

    date_review.append({
        "Column": column,
        "CurrentSourceDtype": str(
            source_df[column].dtype
        ),
        "InvalidDateCount": parsed_dates.isna().sum(),
        "MinimumDate": parsed_dates.min(),
        "MaximumDate": parsed_dates.max()
    })

display(pd.DataFrame(date_review))


# =========================================================
# 13. PaymentDate / CreatedOn / ModifiedOn Date Logic
# =========================================================

print("\n\n13. PAYMENTDATE / CREATEDON / MODIFIEDON DATE LOGIC REVIEW")
print("=" * 80)

payment_dates = pd.to_datetime(
    source_df["PaymentDate"],
    errors="coerce"
)

created_dates = pd.to_datetime(
    source_df["CreatedOn"],
    errors="coerce"
)

modified_dates = pd.to_datetime(
    source_df["ModifiedOn"],
    errors="coerce"
)

payment_before_created = source_df[
    payment_dates < created_dates
]

modified_before_created = source_df[
    modified_dates < created_dates
]

print(
    "Records where PaymentDate is earlier than CreatedOn: "
    f"{len(payment_before_created)}"
)

print(
    "Records where ModifiedOn is earlier than CreatedOn: "
    f"{len(modified_before_created)}"
)

if not payment_before_created.empty:

    print("\nPaymentDate earlier than CreatedOn:")

    display(
        payment_before_created[
            [
                "PaymentReference",
                "PaymentDate",
                "CreatedOn"
            ]
        ]
    )

if not modified_before_created.empty:

    print("\nModifiedOn earlier than CreatedOn:")

    display(
        modified_before_created[
            [
                "PaymentReference",
                "CreatedOn",
                "ModifiedOn"
            ]
        ]
    )


# =========================================================
# 14. Payment Method / Status / Notes Relationship Review
# =========================================================

print("\n\n14. PAYMENT METHOD / STATUS / NOTES RELATIONSHIP REVIEW")
print("=" * 80)

payment_status_notes_review = source_df[
    [
        "PaymentReference",
        "PaymentMethod",
        "PaymentStatus",
        "Amount",
        "PaymentNotes"
    ]
].copy()

display(payment_status_notes_review)


# =========================================================
# 15. Source Datatype Review
# =========================================================

print("\n\n15. SOURCE DATATYPE REVIEW")
print("=" * 80)

datatype_review = pd.DataFrame({
    "Column": source_df.columns,
    "SourceDtype": [
        str(source_df[column].dtype)
        for column in source_df.columns
    ]
})

display(datatype_review)


# =========================================================
# Independent Analyst Review Complete
# =========================================================

print("\n" + "=" * 80)
print("Independent Analyst Review completed.")
print("No data was modified.")
print("=" * 80)
```

## 6.2.1 Create Payment Table
