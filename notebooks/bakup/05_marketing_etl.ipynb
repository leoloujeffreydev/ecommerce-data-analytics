## 1.7 Independent Analyst Review

Perform an independent Pre-ETL-style review of the Marketing Campaign
raw dataset inside the ETL notebook.

This review independently examines the actual dataset values and does
not rely solely on automated profiler findings.

The review covers:

* Dataset structure and grain
* Every source column and its actual values
* Missing values and duplicate records
* Identifier integrity
* Text quality and whitespace
* Categorical consistency
* Numeric validity and reasonableness
* Date values and date relationships
* Boolean values
* Audit field consistency
* Business-rule validation

The purpose is to identify confirmed data-quality issues, distinguish
valid business scenarios from genuine problems, and determine what
requires cleaning, transformation, or an analyst/business decision.

No transformations are applied in this section.

```python
# =========================================================
# 1.7 Independent Analyst Review
# =========================================================

print("Marketing Campaign - Independent Analyst Review")
print("=" * 80)

# ---------------------------------------------------------
# 1. Dataset Overview and Grain
# ---------------------------------------------------------

print("\n1. Dataset Overview and Grain")
print("Expected grain: One marketing campaign per row")
print("Rows:", marketing_campaign_master_raw.shape[0])
print("Columns:", marketing_campaign_master_raw.shape[1])

# ---------------------------------------------------------
# 2. Complete Raw Dataset Review
# ---------------------------------------------------------

print("\n2. Complete Raw Dataset Review")
display(marketing_campaign_master_raw)

# ---------------------------------------------------------
# 3. Missing and Blank Value Review
# ---------------------------------------------------------

print("\n3. Missing and Blank Value Review")

missing_values = marketing_campaign_master_raw.isnull().sum()

blank_values = pd.Series(
    {
        column: (
            marketing_campaign_master_raw[column]
            .astype("string")
            .str.strip()
            .eq("")
            .sum()
            if marketing_campaign_master_raw[column].dtype == "object"
            else 0
        )
        for column in marketing_campaign_master_raw.columns
    }
)

missing_blank_review = pd.DataFrame({
    "MissingValues": missing_values,
    "BlankValues": blank_values
})

display(missing_blank_review)

# ---------------------------------------------------------
# 4. Exact Duplicate Record Review
# ---------------------------------------------------------

print("\n4. Exact Duplicate Record Review")

duplicate_rows = marketing_campaign_master_raw[
    marketing_campaign_master_raw.duplicated(keep=False)
]

print("Exact duplicate rows:", duplicate_rows.shape[0])

if duplicate_rows.empty:
    print("No exact duplicate records found.")
else:
    display(duplicate_rows)

# ---------------------------------------------------------
# 5. CampaignCode Identifier Review
# ---------------------------------------------------------

print("\n5. CampaignCode Identifier Review")

print(
    "Missing CampaignCode values:",
    marketing_campaign_master_raw["CampaignCode"].isnull().sum()
)

print(
    "Duplicate CampaignCode values:",
    marketing_campaign_master_raw["CampaignCode"].duplicated().sum()
)

print("CampaignCode values:")
display(
    marketing_campaign_master_raw[
        ["CampaignCode", "CampaignName"]
    ]
)

# ---------------------------------------------------------
# 6. Campaign Text Field Review
# ---------------------------------------------------------

print("\n6. Campaign Text Field Review")

text_columns = [
    "CampaignName",
    "CampaignType",
    "Description",
    "CreatedByUser",
    "ModifiedByUser"
]

for column in text_columns:

    print(f"\n{column}")

    print(
        "Unique values:",
        marketing_campaign_master_raw[column].nunique()
    )

    column_as_string = (
        marketing_campaign_master_raw[column]
        .astype("string")
    )

    print(
        "Whitespace issues:",
        (
            column_as_string
            != column_as_string.str.strip()
        ).sum()
    )

    print(
        "Blank values:",
        column_as_string.str.strip().eq("").sum()
    )

    print("Actual values:")
    print(marketing_campaign_master_raw[column].tolist())

# ---------------------------------------------------------
# 7. Campaign Type Review
# ---------------------------------------------------------

print("\n7. Campaign Type Review")

print("CampaignType values and frequency:")
print(
    marketing_campaign_master_raw["CampaignType"]
    .value_counts(dropna=False)
)

# ---------------------------------------------------------
# 8. Discount Percentage Review
# ---------------------------------------------------------

print("\n8. Discount Percentage Review")

print("DiscountPct values:")
print(
    marketing_campaign_master_raw["DiscountPct"].tolist()
)

print(
    "\nMinimum:",
    marketing_campaign_master_raw["DiscountPct"].min()
)

print(
    "Maximum:",
    marketing_campaign_master_raw["DiscountPct"].max()
)

print(
    "Negative values:",
    (
        marketing_campaign_master_raw["DiscountPct"] < 0
    ).sum()
)

print(
    "Values above 100:",
    (
        marketing_campaign_master_raw["DiscountPct"] > 100
    ).sum()
)

# ---------------------------------------------------------
# 9. ActiveFlag Boolean Review
# ---------------------------------------------------------

print("\n9. ActiveFlag Boolean Review")

print("ActiveFlag values and frequency:")
print(
    marketing_campaign_master_raw["ActiveFlag"]
    .value_counts(dropna=False)
)

print("\nActual ActiveFlag values:")
print(
    marketing_campaign_master_raw["ActiveFlag"].tolist()
)

# ---------------------------------------------------------
# 10. Date Value Review
# ---------------------------------------------------------

print("\n10. Date Value Review")

date_columns = [
    "StartDate",
    "EndDate",
    "CreatedOn",
    "ModifiedOn"
]

for column in date_columns:

    print(f"\n{column}")

    print("Actual values:")
    print(
        marketing_campaign_master_raw[column].tolist()
    )

    converted_dates = pd.to_datetime(
        marketing_campaign_master_raw[column],
        errors="coerce"
    )

    print(
        "Invalid dates:",
        converted_dates.isnull().sum()
    )

    print(
        "Minimum date:",
        converted_dates.min()
    )

    print(
        "Maximum date:",
        converted_dates.max()
    )

# ---------------------------------------------------------
# 11. Date Business Logic Review
# ---------------------------------------------------------

print("\n11. Date Business Logic Review")

start_date = pd.to_datetime(
    marketing_campaign_master_raw["StartDate"],
    errors="coerce"
)

end_date = pd.to_datetime(
    marketing_campaign_master_raw["EndDate"],
    errors="coerce"
)

created_date = pd.to_datetime(
    marketing_campaign_master_raw["CreatedOn"],
    errors="coerce"
)

modified_date = pd.to_datetime(
    marketing_campaign_master_raw["ModifiedOn"],
    errors="coerce"
)

print(
    "EndDate before StartDate:",
    (end_date < start_date).sum()
)

print(
    "ModifiedOn before CreatedOn:",
    (modified_date < created_date).sum()
)

print(
    "CreatedOn after StartDate:",
    (created_date > start_date).sum()
)

print("\nCampaign Date Relationships:")

display(
    marketing_campaign_master_raw[
        [
            "CampaignCode",
            "CampaignName",
            "StartDate",
            "EndDate",
            "ActiveFlag",
            "CreatedOn",
            "ModifiedOn"
        ]
    ]
)

# ---------------------------------------------------------
# 12. Future Date Review
# ---------------------------------------------------------

print("\n12. Future Date Review")

today = pd.Timestamp.today().normalize()

for column, converted_dates in {
    "StartDate": start_date,
    "EndDate": end_date,
    "CreatedOn": created_date,
    "ModifiedOn": modified_date
}.items():

    future_count = (
        converted_dates > today
    ).sum()

    print(
        f"{column} future dates: {future_count}"
    )

# ---------------------------------------------------------
# 13. Audit Field Review
# ---------------------------------------------------------

print("\n13. Audit Field Review")

print("\nCreatedByUser values:")
print(
    marketing_campaign_master_raw["CreatedByUser"]
    .value_counts(dropna=False)
)

print("\nModifiedByUser values:")
print(
    marketing_campaign_master_raw["ModifiedByUser"]
    .value_counts(dropna=False)
)

# ---------------------------------------------------------
# 14. Independent Review Complete
# ---------------------------------------------------------

print("\n14. Independent Analyst Review Complete")
print("=" * 80)

print(
    "All major Marketing Campaign data areas were independently reviewed."
)

print(
    "No transformations were applied to the raw dataset."
)
```

## 3.2 Data Preservation Validation

Validate that the approved date datatype transformations did not
unintentionally modify the Marketing Campaign dataset.

This validation confirms that:

* Row count remains unchanged.
* Column count remains unchanged.
* Column names remain unchanged.
* Non-date column values remain unchanged.
* Date values remain unchanged after datatype conversion.

Only the approved pandas datatypes should have changed.

```python
# =========================================================
# 3.2 Data Preservation Validation
# =========================================================

print("Marketing Campaign Data Preservation Validation")
print("=" * 80)

# ---------------------------------------------------------
# 1. Row Count Validation
# ---------------------------------------------------------

print("\n1. Row Count Validation")

raw_row_count = marketing_campaign_master_raw.shape[0]
clean_row_count = marketing_campaign_master_clean.shape[0]

print("Raw dataset rows:", raw_row_count)
print("Clean dataset rows:", clean_row_count)
print("Row counts match:", raw_row_count == clean_row_count)

# ---------------------------------------------------------
# 2. Column Structure Validation
# ---------------------------------------------------------

print("\n2. Column Structure Validation")

raw_columns = marketing_campaign_master_raw.columns.tolist()
clean_columns = marketing_campaign_master_clean.columns.tolist()

print("Column counts match:",
      len(raw_columns) == len(clean_columns))

print("Column names and order match:",
      raw_columns == clean_columns)

# ---------------------------------------------------------
# 3. Non-Date Value Preservation
# ---------------------------------------------------------

print("\n3. Non-Date Value Preservation")

non_date_columns = [
    column
    for column in marketing_campaign_master_raw.columns
    if column not in date_columns
]

non_date_values_match = (
    marketing_campaign_master_raw[non_date_columns]
    .equals(marketing_campaign_master_clean[non_date_columns])
)

print("Non-date values unchanged:",
      non_date_values_match)

# ---------------------------------------------------------
# 4. Date Value Preservation
# ---------------------------------------------------------

print("\n4. Date Value Preservation")

date_values_match = True

for column in date_columns:

    raw_dates = pd.to_datetime(
        marketing_campaign_master_raw[column],
        errors="raise"
    )

    clean_dates = marketing_campaign_master_clean[column]

    values_match = raw_dates.equals(clean_dates)

    print(f"{column:<15} Values unchanged: {values_match}")

    if not values_match:
        date_values_match = False

# ---------------------------------------------------------
# 5. Overall Preservation Validation
# ---------------------------------------------------------

all_preservation_checks_passed = (
    raw_row_count == clean_row_count
    and raw_columns == clean_columns
    and non_date_values_match
    and date_values_match
)

print("\n5. Data Preservation Validation Summary")
print("=" * 80)

print(
    "All data preservation checks passed:",
    all_preservation_checks_passed
)
```

- Duplicate record checks
- Boolean validity
- Campaign date consistency

These checks provide the final confirmation that the Marketing Campaign

```python
# =========================================================
# 6.9 Marketing Campaign Database Integrity Validation
# =========================================================

integrity_query = """
SELECT
    COUNT(*) AS total_rows,

    COUNT(DISTINCT campaign_id) AS distinct_campaign_ids,

    COUNT(*) FILTER (
        WHERE campaign_id IS NULL
    ) AS null_campaign_ids,

    COUNT(*) FILTER (
        WHERE campaign_name IS NULL
    ) AS null_campaign_names,

    COUNT(*) FILTER (
        WHERE campaign_type IS NULL
    ) AS null_campaign_types,

    COUNT(*) FILTER (
        WHERE start_date IS NULL
    ) AS null_start_dates,

    COUNT(*) FILTER (
        WHERE end_date IS NULL
    ) AS null_end_dates,

    COUNT(*) FILTER (
        WHERE discount_percentage IS NULL
    ) AS null_discount_percentages,

    COUNT(*) FILTER (
        WHERE is_active IS NULL
    ) AS null_active_flags,

    COUNT(*) FILTER (
        WHERE created_date IS NULL
    ) AS null_created_dates,

    COUNT(*) FILTER (
        WHERE created_by IS NULL
    ) AS null_created_by,

    COUNT(*) FILTER (
        WHERE updated_date IS NULL
    ) AS null_updated_dates,

    COUNT(*) FILTER (
        WHERE updated_by IS NULL
    ) AS null_updated_by,

    COUNT(*) FILTER (
        WHERE start_date > end_date
    ) AS invalid_campaign_date_order,

    COUNT(*) FILTER (
        WHERE updated_date < created_date
    ) AS invalid_audit_date_order

FROM commerce.marketing_campaign;
"""

try:

    cursor.execute(integrity_query)

    integrity_result = cursor.fetchone()

    print("Marketing Campaign Database Integrity Validation")
    print("=" * 80)

    print("\n1. Primary Key Integrity")
    print("Total rows:", integrity_result[0])
    print("Distinct campaign IDs:", integrity_result[1])
    print("NULL campaign IDs:", integrity_result[2])

    print("\n2. Required Field Integrity")
    print("NULL campaign names:", integrity_result[3])
    print("NULL campaign types:", integrity_result[4])
    print("NULL start dates:", integrity_result[5])
    print("NULL end dates:", integrity_result[6])
    print("NULL discount percentages:", integrity_result[7])
    print("NULL active flags:", integrity_result[8])
    print("NULL created dates:", integrity_result[9])
    print("NULL created by:", integrity_result[10])
    print("NULL updated dates:", integrity_result[11])
    print("NULL updated by:", integrity_result[12])

    print("\n3. Campaign Date Integrity")
    print(
        "Start date later than end date:",
        integrity_result[13]
    )

    print("\n4. Audit Date Integrity")
    print(
        "Updated date earlier than created date:",
        integrity_result[14]
    )

    # -----------------------------------------------------
    # 5. Exact Duplicate Record Check
    # -----------------------------------------------------

    cursor.execute("""
        SELECT COUNT(*)
        FROM (
            SELECT
                campaign_id,
                campaign_name,
                campaign_type,
                start_date,
                end_date,
                discount_percentage,
                campaign_description,
                is_active,
                created_date,
                created_by,
                updated_date,
                updated_by
            FROM commerce.marketing_campaign
            GROUP BY
                campaign_id,
                campaign_name,
                campaign_type,
                start_date,
                end_date,
                discount_percentage,
                campaign_description,
                is_active,
                created_date,
                created_by,
                updated_date,
                updated_by
            HAVING COUNT(*) > 1
        ) AS duplicate_groups;
    """)

    duplicate_record_groups = cursor.fetchone()[0]

    print("\n5. Exact Duplicate Record Check")
    print(
        "Duplicate record groups:",
        duplicate_record_groups
    )

    # -----------------------------------------------------
    # 6. Boolean Validity
    # -----------------------------------------------------

    cursor.execute("""
        SELECT COUNT(*)
        FROM commerce.marketing_campaign
        WHERE is_active NOT IN (TRUE, FALSE);
    """)

    invalid_boolean_count = cursor.fetchone()[0]

    print("\n6. Boolean Validity")
    print(
        "Invalid ActiveFlag values:",
        invalid_boolean_count
    )

    # -----------------------------------------------------
    # Final Validation Result
    # -----------------------------------------------------

    integrity_passed = (
        integrity_result[0] == integrity_result[1]
        and integrity_result[2] == 0
        and integrity_result[3] == 0
        and integrity_result[4] == 0
        and integrity_result[5] == 0
        and integrity_result[6] == 0
        and integrity_result[7] == 0
        and integrity_result[8] == 0
        and integrity_result[9] == 0
        and integrity_result[10] == 0
        and integrity_result[11] == 0
        and integrity_result[12] == 0
        and integrity_result[13] == 0
        and integrity_result[14] == 0
        and duplicate_record_groups == 0
        and invalid_boolean_count == 0
    )

    print("\n7. Database Integrity Validation Summary")
    print("=" * 80)

    print(
        "Marketing Campaign database integrity passed:",
        integrity_passed
    )

except Exception as e:

    print(
        "Marketing Campaign database integrity validation failed."
    )

    print("Error:", e)

    raise
```
