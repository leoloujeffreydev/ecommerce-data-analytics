## 1.7 Independent Analyst Review

Perform a systematic independent analyst review of the untouched raw Sales Order datasets.

This section is the notebook-based version of the Pre-ETL Inspection.

The review goes beyond the automated profiler and checks each Sales Order source dataset using independent analyst validation relevant to its business meaning and approved ETL design.

The review covers, where applicable:

- Dataset structure and grain
- Missing and blank values
- Exact duplicate rows
- Primary-key candidates and key uniqueness
- Cross-source key uniqueness and collision checks
- Whitespace in text fields
- Text consistency
- Numeric validity and statistical outlier candidates
- Date parsing, datatype, validity, and Created/Modified date consistency
- CustomerCode and CampaignCode relationship values
- Source-to-target datatype compatibility considerations
- Cross-source structural consistency

Automated profiler findings and independent analyst findings are kept conceptually separate.

Each source dataset is reviewed independently before consolidation.

No data is modified in this section.

```python
# =========================================================
# 1.7 Independent Analyst Review
# =========================================================


# =========================================================
# 1. Dataset Structure and Grain
# =========================================================

print("1. DATASET STRUCTURE AND GRAIN")
print("=" * 80)

sales_order_sources = {
    "Shopify": shopify_orders_raw,
    "Amazon UAE": amazon_orders_raw,
    "Noon": noon_orders_raw
}

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    print(f"Rows: {source_df.shape[0]}")
    print(f"Columns: {source_df.shape[1]}")

    print("\nWorking assumption:")
    print("One row represents one customer order / sales order header.")


# =========================================================
# 2. Missing Values and Blank Values
# =========================================================

print("\n\n2. MISSING VALUES AND BLANK VALUES")
print("=" * 80)

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    missing_values = source_df.isna().sum()

    blank_values = pd.Series(
        {
            column: (
                source_df[column]
                .astype("string")
                .str.strip()
                .eq("")
                .sum()
                if source_df[column].dtype == "object"
                else 0
            )
            for column in source_df.columns
        }
    )

    missing_blank_review = pd.DataFrame({
        "MissingValues": missing_values,
        "BlankValues": blank_values
    })

    display(missing_blank_review)


# =========================================================
# 3. Exact Duplicate Rows
# =========================================================

print("\n\n3. EXACT DUPLICATE ROWS")
print("=" * 80)

for source_name, source_df in sales_order_sources.items():

    exact_duplicate_count = source_df.duplicated().sum()

    print(
        f"{source_name} - Exact duplicate rows: "
        f"{exact_duplicate_count}"
    )

    if exact_duplicate_count > 0:

        display(
            source_df[
                source_df.duplicated(keep=False)
            ]
        )


# =========================================================
# 4. Primary Key Candidate and Identifier Uniqueness
# =========================================================

print("\n\n4. PRIMARY KEY CANDIDATE AND IDENTIFIER UNIQUENESS")
print("=" * 80)

identifier_columns = [
    "SourceOrderID",
    "SourceOrderIDInternal",
    "CustomerCode",
    "CampaignCode"
]

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    identifier_review = pd.DataFrame(
        {
            "Column": identifier_columns,
            "NullCount": [
                source_df[column].isna().sum()
                for column in identifier_columns
            ],
            "BlankCount": [
                source_df[column]
                .astype("string")
                .str.strip()
                .eq("")
                .sum()
                for column in identifier_columns
            ],
            "DuplicateCount": [
                source_df[column]
                .duplicated()
                .sum()
                for column in identifier_columns
            ],
            "UniqueValues": [
                source_df[column]
                .nunique()
                for column in identifier_columns
            ]
        }
    )

    display(identifier_review)


# =========================================================
# 5. Cross-Source Order Identifier Collision Review
# =========================================================

print("\n\n5. CROSS-SOURCE ORDER IDENTIFIER COLLISION REVIEW")
print("=" * 80)

cross_source_ids = pd.concat(
    [
        shopify_orders_raw[
            ["SourceSystem", "SourceOrderID", "SourceOrderIDInternal"]
        ],
        amazon_orders_raw[
            ["SourceSystem", "SourceOrderID", "SourceOrderIDInternal"]
        ],
        noon_orders_raw[
            ["SourceSystem", "SourceOrderID", "SourceOrderIDInternal"]
        ]
    ],
    ignore_index=True
)

print("Total consolidated source-order records:")
print(len(cross_source_ids))

print("\nUnique SourceOrderIDInternal values:")
print(cross_source_ids["SourceOrderIDInternal"].nunique())

cross_source_internal_duplicates = (
    cross_source_ids[
        cross_source_ids["SourceOrderIDInternal"].duplicated(
            keep=False
        )
    ]
    .sort_values("SourceOrderIDInternal")
)

print(
    "\nCross-source SourceOrderIDInternal collisions: "
    f"{len(cross_source_internal_duplicates)}"
)

if not cross_source_internal_duplicates.empty:

    display(cross_source_internal_duplicates)


# =========================================================
# 6. Leading and Trailing Whitespace Review
# =========================================================

print("\n\n6. LEADING AND TRAILING WHITESPACE REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    text_columns = source_df.select_dtypes(
        include="object"
    ).columns

    whitespace_findings = []

    for column in text_columns:

        whitespace_mask = (
            source_df[column]
            .astype("string")
            != source_df[column]
            .astype("string")
            .str.strip()
        )

        count = whitespace_mask.sum()

        if count > 0:

            whitespace_records = source_df.loc[
                whitespace_mask,
                ["SourceOrderID", column]
            ]

            whitespace_findings.append({
                "Column": column,
                "WhitespaceRecordCount": count
            })

            print(f"\nColumn: {column}")
            display(whitespace_records)

    if whitespace_findings:

        print("\nWhitespace findings summary:")
        display(pd.DataFrame(whitespace_findings))

    else:

        print("No leading or trailing whitespace found.")


# =========================================================
# 7. Text Consistency Review
# =========================================================

print("\n\n7. TEXT CONSISTENCY REVIEW")
print("=" * 80)

text_consistency_columns = [
    "SourceSystem",
    "OrderStatus",
    "PaymentMethod"
]

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    for column in text_consistency_columns:

        print(f"\n{column} values:")

        display(
            source_df[column]
            .value_counts(dropna=False)
            .sort_index()
            .rename("RecordCount")
            .to_frame()
        )


# =========================================================
# 8. Numeric Validity Review
# =========================================================

print("\n\n8. NUMERIC VALIDITY REVIEW")
print("=" * 80)

numeric_columns = [
    "OrderGrossAmount"
]

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    numeric_summary = source_df[
        numeric_columns
    ].describe().T

    numeric_summary["NegativeCount"] = [
        (source_df[column] < 0).sum()
        for column in numeric_columns
    ]

    numeric_summary["ZeroCount"] = [
        (source_df[column] == 0).sum()
        for column in numeric_columns
    ]

    numeric_summary["PositiveCount"] = [
        (source_df[column] > 0).sum()
        for column in numeric_columns
    ]

    display(numeric_summary)


# =========================================================
# 9. Numeric Outlier Candidate Review
# =========================================================

print("\n\n9. NUMERIC OUTLIER CANDIDATE REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    outlier_summary = []

    for column in numeric_columns:

        q1 = source_df[column].quantile(0.25)
        q3 = source_df[column].quantile(0.75)

        iqr = q3 - q1

        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)

        outlier_mask = (
            (source_df[column] < lower_bound)
            |
            (source_df[column] > upper_bound)
        )

        outlier_count = outlier_mask.sum()

        outlier_summary.append({
            "Column": column,
            "LowerBoundary": lower_bound,
            "UpperBoundary": upper_bound,
            "OutlierCandidateCount": outlier_count
        })

        if outlier_count > 0:

            print(f"\nOutlier candidate(s): {column}")

            display(
                source_df.loc[
                    outlier_mask,
                    [
                        "SourceOrderID",
                        "SourceOrderIDInternal",
                        "CustomerCode",
                        column
                    ]
                ]
            )

    print("\nOutlier candidate summary:")
    display(pd.DataFrame(outlier_summary))


# =========================================================
# 10. Date Parsing and Datatype Review
# =========================================================

print("\n\n10. DATE PARSING AND DATATYPE REVIEW")
print("=" * 80)

date_columns = [
    "OrderDate",
    "CreatedOn",
    "ModifiedOn"
]

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    date_review = []

    for column in date_columns:

        parsed_dates = pd.to_datetime(
            source_df[column],
            errors="coerce"
        )

        date_review.append({
            "Column": column,
            "CurrentSourceDtype": str(
                source_df[column].dtype
            ),
            "InvalidDateCount": parsed_dates.isna().sum(),
            "MinimumDate": parsed_dates.min(),
            "MaximumDate": parsed_dates.max()
        })

    display(pd.DataFrame(date_review))


# =========================================================
# 11. CreatedOn / ModifiedOn and OrderDate Logic
# =========================================================

print("\n\n11. DATE LOGIC REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    order_dates = pd.to_datetime(
        source_df["OrderDate"],
        errors="coerce"
    )

    created_dates = pd.to_datetime(
        source_df["CreatedOn"],
        errors="coerce"
    )

    modified_dates = pd.to_datetime(
        source_df["ModifiedOn"],
        errors="coerce"
    )

    modified_before_created = source_df[
        modified_dates < created_dates
    ]

    created_before_order = source_df[
        created_dates < order_dates
    ]

    modified_before_order = source_df[
        modified_dates < order_dates
    ]

    print(
        "Records where ModifiedOn is earlier than CreatedOn: "
        f"{len(modified_before_created)}"
    )

    print(
        "Records where CreatedOn is earlier than OrderDate: "
        f"{len(created_before_order)}"
    )

    print(
        "Records where ModifiedOn is earlier than OrderDate: "
        f"{len(modified_before_order)}"
    )

    if not modified_before_created.empty:

        print("\nModifiedOn earlier than CreatedOn:")

        display(
            modified_before_created[
                [
                    "SourceOrderID",
                    "OrderDate",
                    "CreatedOn",
                    "ModifiedOn"
                ]
            ]
        )


# =========================================================
# 12. CustomerCode and CampaignCode Relationship Review
# =========================================================

print("\n\n12. RELATIONSHIP VALUE REVIEW")
print("=" * 80)

relationship_columns = [
    "CustomerCode",
    "CampaignCode"
]

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    for column in relationship_columns:

        print(f"\n{column} frequency:")

        display(
            source_df[column]
            .value_counts(dropna=False)
            .sort_index()
            .rename("RecordCount")
            .to_frame()
        )


# =========================================================
# 13. Sales Order and Campaign Date Relationship Review
# =========================================================

print("\n\n13. SALES ORDER AND CAMPAIGN DATE RELATIONSHIP REVIEW")
print("=" * 80)

print(
    "Review Sales Order records where OrderDate occurs before "
    "the associated Campaign StartDate."
)

# ---------------------------------------------------------
# Load Marketing Campaign raw dataset
# ---------------------------------------------------------

marketing_campaign_raw_path = Path(
    r"C:\JEP\DATA ANALYST PORTFOLIO\Lj_Dev_Commerce"
    r"\data\05_Marketing\marketing_campaign_master.csv"
)

marketing_campaign_raw = pd.read_csv(
    marketing_campaign_raw_path
)

# ---------------------------------------------------------
# Combine Sales Order raw datasets
# ---------------------------------------------------------

sales_order_campaign_review = pd.concat(
    [
        shopify_orders_raw,
        amazon_orders_raw,
        noon_orders_raw
    ],
    ignore_index=True
)

# ---------------------------------------------------------
# Convert dates for comparison
# ---------------------------------------------------------

sales_order_campaign_review["OrderDate_Parsed"] = pd.to_datetime(
    sales_order_campaign_review["OrderDate"],
    errors="coerce"
)

marketing_campaign_raw_review = (
    marketing_campaign_raw.copy(deep=True)
)

marketing_campaign_raw_review["StartDate_Parsed"] = pd.to_datetime(
    marketing_campaign_raw_review["StartDate"],
    errors="coerce"
)

# ---------------------------------------------------------
# Match Sales Orders to Marketing Campaigns
# ---------------------------------------------------------

campaign_date_review = sales_order_campaign_review.merge(
    marketing_campaign_raw_review[
        [
            "CampaignCode",
            "StartDate_Parsed"
        ]
    ],
    on="CampaignCode",
    how="left"
)

# ---------------------------------------------------------
# Identify campaign date exceptions
# ---------------------------------------------------------

campaign_date_exception_mask = (
    campaign_date_review["CampaignCode"].notna()
    &
    campaign_date_review["OrderDate_Parsed"].notna()
    &
    campaign_date_review["StartDate_Parsed"].notna()
    &
    (
        campaign_date_review["OrderDate_Parsed"]
        < campaign_date_review["StartDate_Parsed"]
    )
)

campaign_date_exceptions = campaign_date_review.loc[
    campaign_date_exception_mask,
    [
        "SourceSystem",
        "SourceOrderID",
        "SourceOrderIDInternal",
        "CustomerCode",
        "CampaignCode",
        "OrderDate",
        "StartDate_Parsed"
    ]
].copy()

print(
    "Sales Order records with OrderDate before Campaign StartDate:",
    len(campaign_date_exceptions)
)

if not campaign_date_exceptions.empty:

    display(campaign_date_exceptions)

else:

    print("No campaign date relationship exceptions found.")


# =========================================================
# 14. Source Datatype Review
# =========================================================

print("\n\n14. SOURCE DATATYPE REVIEW")
print("=" * 80)

for source_name, source_df in sales_order_sources.items():

    print(f"\n{source_name}")
    print("-" * 80)

    datatype_review = pd.DataFrame({
        "Column": source_df.columns,
        "SourceDtype": [
            str(source_df[column].dtype)
            for column in source_df.columns
        ]
    })

    display(datatype_review)


# =========================================================
# 15. Cross-Source Structural Consistency Review
# =========================================================

print("\n\n15. CROSS-SOURCE STRUCTURAL CONSISTENCY REVIEW")
print("=" * 80)

source_structure_review = pd.DataFrame(
    {
        "Source": list(sales_order_sources.keys()),
        "Rows": [
            source_df.shape[0]
            for source_df in sales_order_sources.values()
        ],
        "Columns": [
            source_df.shape[1]
            for source_df in sales_order_sources.values()
        ],
        "ColumnOrderMatchesShopify": [
            list(source_df.columns)
            == list(shopify_orders_raw.columns)
            for source_df in sales_order_sources.values()
        ]
    }
)

display(source_structure_review)

print("\nExpected Sales Order source columns:")
print(shopify_orders_raw.columns.tolist())


# =========================================================
# Independent Analyst Review Complete
# =========================================================

print("\n" + "=" * 80)
print("Independent Analyst Review completed.")
print("No data was modified.")
print("=" * 80)
```

## 2.2 Apply Approved Transformations

Apply only the transformations approved in Section 1.8 to the clean working Sales Order datasets.

Approved transformations:

- Trim leading and trailing whitespace from OrderStatus.
- Standardize known OrderStatus values to the approved canonical values: Completed, Pending, Shipped, and Cancelled.
- Convert OrderDate to datetime.
- Convert CreatedOn to datetime.
- Convert ModifiedOn to datetime.

The OrderStatus standardization is deterministic and ensures equivalent source values are represented consistently for filtering, grouping, reporting, and downstream BI analysis.

No other business values will be modified.

The Amazon SO013 statistical outlier candidate will be retained unchanged because statistical outlier detection alone does not confirm a data error.

The SO005 and SO007 CAM002 temporal exceptions will be retained unchanged because there is insufficient evidence to determine the correct business value.

The SO007 Sales Order total and payment discrepancy will also be retained unchanged because the available data does not establish the reason for the AED 179 difference.

The three source datasets will remain separate during this transformation step.

```python
# =========================================================
# 2.2 Apply Approved Transformations
# =========================================================

# ---------------------------------------------------------
# 1. Standardize approved OrderStatus values
# ---------------------------------------------------------

status_standardization = {
    "completed": "Completed",
    "pending": "Pending",
    "shipped": "Shipped",
    "cancelled": "Cancelled"
}

for df in [
    shopify_orders_clean,
    amazon_orders_clean,
    noon_orders_clean
]:
    df["OrderStatus"] = (
        df["OrderStatus"]
        .str.strip()
        .str.lower()
        .replace(status_standardization)
    )


# ---------------------------------------------------------
# 2. Convert approved date fields
# ---------------------------------------------------------

for df in [
    shopify_orders_clean,
    amazon_orders_clean,
    noon_orders_clean
]:
    df["OrderDate"] = pd.to_datetime(df["OrderDate"], errors="raise")
    df["CreatedOn"] = pd.to_datetime(df["CreatedOn"], errors="raise")
    df["ModifiedOn"] = pd.to_datetime(df["ModifiedOn"], errors="raise")


# ---------------------------------------------------------
# 3. Confirm approved OrderStatus values only
# ---------------------------------------------------------

approved_order_statuses = {
    "Completed",
    "Pending",
    "Shipped",
    "Cancelled"
}

for source_name, df in [
    ("Shopify", shopify_orders_clean),
    ("Amazon UAE", amazon_orders_clean),
    ("Noon", noon_orders_clean)
]:
    unexpected_statuses = sorted(
        set(df["OrderStatus"].dropna()) - approved_order_statuses
    )

    if unexpected_statuses:
        raise ValueError(
            f"{source_name} contains unexpected OrderStatus values "
            f"after standardization: {unexpected_statuses}"
        )


print("Sales Order approved transformations applied successfully.")
    
```

Consolidate the validated Clean DataFrames from Shopify, Amazon UAE,

## 6.2.1 Create Sales Order Table

The `commerce.sales_order` table does not currently exist.

This section creates the Sales Order PostgreSQL target table based

on the approved LJ Dev Commerce Sales Order Data Dictionary.

Key design decisions:

- `sales_order_id` is the Primary Key.

- All required Data Dictionary fields use `NOT NULL`.

- `campaign_id`, `shipping_address`, and `notes` are optional and

  therefore allow `NULL`.

- `customer_id` and `campaign_id` are Foreign Keys based on the

  approved Sales Order database relationships.

- `order_date`, `created_date`, and `updated_date` use the PostgreSQL

  `TIMESTAMP` datatype.

- `order_total` uses the PostgreSQL `DECIMAL` datatype.

```python
# =========================================================
# 6.2.1 Create Sales Order Table
# =========================================================

create_sales_order_table_query = """
CREATE TABLE commerce.sales_order (

    sales_order_id TEXT PRIMARY KEY,

    customer_id TEXT NOT NULL,
    campaign_id TEXT,

    order_date TIMESTAMP NOT NULL,
    order_status TEXT NOT NULL,

    shipping_address TEXT,
    billing_address TEXT NOT NULL,

    payment_method TEXT NOT NULL,
    order_total DECIMAL NOT NULL,

    notes TEXT,

    created_date TIMESTAMP NOT NULL,
    created_by TEXT NOT NULL,

    updated_date TIMESTAMP NOT NULL,
    updated_by TEXT NOT NULL,

    FOREIGN KEY (customer_id)
        REFERENCES commerce.customer(customer_id),

    FOREIGN KEY (campaign_id)
        REFERENCES commerce.marketing_campaign(campaign_id)

);
"""

try:
    cursor.execute(create_sales_order_table_query)
    conn.commit()

    print("Sales Order table created successfully.")
    print("Table: commerce.sales_order")

except Exception as e:
    conn.rollback()

    print("Sales Order table creation failed.")
    print("Error:", e)
```

## 6.9 Sales Order Database Integrity Validation

Perform final integrity checks on the loaded

`commerce.sales_order` table.

The checks confirm:

- Primary-key uniqueness

- Required-field integrity

- Duplicate record checks

- Order status validity

- Order total integrity

- Audit date consistency

These checks provide the final confirmation that the Sales Order

table is structurally and business-rule compliant after loading.

```python
# =========================================================
# 6.9 Sales Order Database Integrity Validation
# =========================================================

integrity_query = """
SELECT
    COUNT(*) AS total_rows,

    COUNT(DISTINCT sales_order_id) AS distinct_sales_order_ids,

    COUNT(*) FILTER (
        WHERE sales_order_id IS NULL
    ) AS null_sales_order_ids,

    COUNT(*) FILTER (
        WHERE customer_id IS NULL
    ) AS null_customer_ids,

    COUNT(*) FILTER (
        WHERE order_date IS NULL
    ) AS null_order_dates,

    COUNT(*) FILTER (
        WHERE order_status IS NULL
    ) AS null_order_statuses,

    COUNT(*) FILTER (
        WHERE billing_address IS NULL
    ) AS null_billing_addresses,

    COUNT(*) FILTER (
        WHERE payment_method IS NULL
    ) AS null_payment_methods,

    COUNT(*) FILTER (
        WHERE order_total IS NULL
    ) AS null_order_totals,

    COUNT(*) FILTER (
        WHERE created_date IS NULL
    ) AS null_created_dates,

    COUNT(*) FILTER (
        WHERE created_by IS NULL
    ) AS null_created_by,

    COUNT(*) FILTER (
        WHERE updated_date IS NULL
    ) AS null_updated_dates,

    COUNT(*) FILTER (
        WHERE updated_by IS NULL
    ) AS null_updated_by,

    COUNT(*) FILTER (
        WHERE order_status NOT IN (
            'Completed',
            'Pending',
            'Shipped',
            'Cancelled'
        )
    ) AS invalid_order_statuses,

    COUNT(*) FILTER (
        WHERE order_total <= 0
    ) AS invalid_order_totals,

    COUNT(*) FILTER (
        WHERE updated_date < created_date
    ) AS invalid_audit_date_order

FROM commerce.sales_order;
"""

try:
    cursor.execute(integrity_query)

    integrity_result = cursor.fetchone()

    print("Sales Order Database Integrity Validation")
    print("=" * 80)

    print("\n1. Primary Key Integrity")
    print("Total rows:", integrity_result[0])
    print("Distinct sales order IDs:", integrity_result[1])
    print("NULL sales order IDs:", integrity_result[2])

    print("\n2. Required Field Integrity")
    print("NULL customer IDs:", integrity_result[3])
    print("NULL order dates:", integrity_result[4])
    print("NULL order statuses:", integrity_result[5])
    print("NULL billing addresses:", integrity_result[6])
    print("NULL payment methods:", integrity_result[7])
    print("NULL order totals:", integrity_result[8])
    print("NULL created dates:", integrity_result[9])
    print("NULL created by:", integrity_result[10])
    print("NULL updated dates:", integrity_result[11])
    print("NULL updated by:", integrity_result[12])

    print("\n3. Sales Order Value Integrity")
    print("Invalid order statuses:", integrity_result[13])
    print("Invalid order totals:", integrity_result[14])

    print("\n4. Audit Date Integrity")
    print(
        "Updated date earlier than created date:",
        integrity_result[15]
    )

    # -----------------------------------------------------
    # Final Validation Result
    # -----------------------------------------------------

    integrity_passed = (
        integrity_result[0] == integrity_result[1]
        and integrity_result[2] == 0
        and integrity_result[3] == 0
        and integrity_result[4] == 0
        and integrity_result[5] == 0
        and integrity_result[6] == 0
        and integrity_result[7] == 0
        and integrity_result[8] == 0
        and integrity_result[9] == 0
        and integrity_result[10] == 0
        and integrity_result[11] == 0
        and integrity_result[12] == 0
        and integrity_result[13] == 0
        and integrity_result[14] == 0
        and integrity_result[15] == 0
    )

    print("\n5. Database Integrity Validation Summary")
    print("=" * 80)

    print(
        "Sales Order database integrity passed:",
        integrity_passed
    )

except Exception as e:
    print("Sales Order database integrity validation failed.")
    print("Error:", e)
```
