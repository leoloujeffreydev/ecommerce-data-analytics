## 1.7 Independent Analyst Review

Perform a systematic independent analyst review of the untouched raw Product dataset.

This section is the notebook-based version of the Pre-ETL Inspection.

The review goes beyond the automated profiler and checks the Product dataset using independent analyst validation relevant to its business meaning and approved ETL design.

The review covers, where applicable:

- Dataset structure and grain
- Missing and blank values
- Exact duplicate rows
- Primary-key candidates and key uniqueness
- Whitespace in text fields
- Text consistency
- Numeric validity and statistical outlier candidates
- Boolean validity
- Date parsing, datatype, validity, and Created/Modified date consistency
- CategoryCode and SupplierCode relationship values
- Source-to-target datatype compatibility considerations

Automated profiler findings and independent analyst findings are kept conceptually separate.

No data is modified in this section.

```python
# =========================================================
# 1.7 Independent Analyst Review
# =========================================================


# =========================================================
# 1. Dataset Structure and Grain
# =========================================================

print("1. DATASET STRUCTURE AND GRAIN")
print("=" * 80)

print(f"Rows: {product_catalog_products_raw.shape[0]}")
print(f"Columns: {product_catalog_products_raw.shape[1]}")

print("\nWorking assumption:")
print("One row represents one product record.")


# =========================================================
# 2. Missing Values and Blank Values
# =========================================================

print("\n2. MISSING VALUES AND BLANK VALUES")
print("=" * 80)

missing_values = product_catalog_products_raw.isna().sum()

blank_values = pd.Series(
    {
        column: (
            product_catalog_products_raw[column]
            .astype("string")
            .str.strip()
            .eq("")
            .sum()
            if product_catalog_products_raw[column].dtype == "object"
            else 0
        )
        for column in product_catalog_products_raw.columns
    }
)

missing_blank_review = pd.DataFrame({
    "MissingValues": missing_values,
    "BlankValues": blank_values
})

display(missing_blank_review)


# =========================================================
# 3. Exact Duplicate Rows
# =========================================================

print("\n3. EXACT DUPLICATE ROWS")
print("=" * 80)

exact_duplicate_count = product_catalog_products_raw.duplicated().sum()

print(f"Exact duplicate rows: {exact_duplicate_count}")

if exact_duplicate_count > 0:
    display(
        product_catalog_products_raw[
            product_catalog_products_raw.duplicated(keep=False)
        ]
    )


# =========================================================
# 4. Primary Key Candidate and Identifier Uniqueness
# =========================================================

print("\n4. PRIMARY KEY CANDIDATE AND IDENTIFIER UNIQUENESS")
print("=" * 80)

identifier_columns = ["ProductCode", "SKU"]

identifier_review = pd.DataFrame(
    {
        "Column": identifier_columns,
        "NullCount": [
            product_catalog_products_raw[column].isna().sum()
            for column in identifier_columns
        ],
        "BlankCount": [
            product_catalog_products_raw[column]
            .astype("string")
            .str.strip()
            .eq("")
            .sum()
            for column in identifier_columns
        ],
        "DuplicateCount": [
            product_catalog_products_raw[column]
            .duplicated()
            .sum()
            for column in identifier_columns
        ],
        "UniqueValues": [
            product_catalog_products_raw[column]
            .nunique()
            for column in identifier_columns
        ]
    }
)

display(identifier_review)


# =========================================================
# 5. Leading and Trailing Whitespace Review
# =========================================================

print("\n5. LEADING AND TRAILING WHITESPACE REVIEW")
print("=" * 80)

text_columns = product_catalog_products_raw.select_dtypes(
    include="object"
).columns

whitespace_findings = []

for column in text_columns:

    whitespace_mask = (
        product_catalog_products_raw[column]
        .astype("string")
        != product_catalog_products_raw[column]
        .astype("string")
        .str.strip()
    )

    count = whitespace_mask.sum()

    if count > 0:

        whitespace_records = product_catalog_products_raw.loc[
            whitespace_mask,
            ["ProductCode", column]
        ]

        whitespace_findings.append({
            "Column": column,
            "WhitespaceRecordCount": count
        })

        print(f"\nColumn: {column}")
        display(whitespace_records)


if whitespace_findings:

    print("\nWhitespace findings summary:")
    display(pd.DataFrame(whitespace_findings))

else:

    print("No leading or trailing whitespace found.")


# =========================================================
# 6. Numeric Validity Review
# =========================================================

print("\n6. NUMERIC VALIDITY REVIEW")
print("=" * 80)

numeric_columns = [
    "UnitCost",
    "ListPrice",
    "WeightKg",
    "WarrantyMonths"
]

numeric_summary = product_catalog_products_raw[
    numeric_columns
].describe().T

numeric_summary["NegativeCount"] = [
    (product_catalog_products_raw[column] < 0).sum()
    for column in numeric_columns
]

display(numeric_summary)


# =========================================================
# 7. Numeric Outlier Candidate Review
# =========================================================

print("\n7. NUMERIC OUTLIER CANDIDATE REVIEW")
print("=" * 80)

outlier_summary = []

for column in numeric_columns:

    q1 = product_catalog_products_raw[column].quantile(0.25)
    q3 = product_catalog_products_raw[column].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)

    outlier_mask = (
        (product_catalog_products_raw[column] < lower_bound)
        |
        (product_catalog_products_raw[column] > upper_bound)
    )

    outlier_count = outlier_mask.sum()

    outlier_summary.append({
        "Column": column,
        "LowerBoundary": lower_bound,
        "UpperBoundary": upper_bound,
        "OutlierCandidateCount": outlier_count
    })

    if outlier_count > 0:

        print(f"\nOutlier candidate(s): {column}")

        display(
            product_catalog_products_raw.loc[
                outlier_mask,
                [
                    "ProductCode",
                    "SKU",
                    "ProductName",
                    column
                ]
            ]
        )


print("\nOutlier candidate summary:")
display(pd.DataFrame(outlier_summary))


# =========================================================
# 8. Boolean Validity Review
# =========================================================

print("\n8. BOOLEAN VALIDITY REVIEW")
print("=" * 80)

print("ActiveFlag datatype:")
print(product_catalog_products_raw["ActiveFlag"].dtype)

print("\nActiveFlag values:")
print(
    product_catalog_products_raw["ActiveFlag"]
    .value_counts(dropna=False)
)


# =========================================================
# 9. Date Parsing and Datatype Review
# =========================================================

print("\n9. DATE PARSING AND DATATYPE REVIEW")
print("=" * 80)

date_columns = ["CreatedOn", "ModifiedOn"]

date_review = []

for column in date_columns:

    parsed_dates = pd.to_datetime(
        product_catalog_products_raw[column],
        errors="coerce"
    )

    date_review.append({
        "Column": column,
        "CurrentSourceDtype": str(
            product_catalog_products_raw[column].dtype
        ),
        "InvalidDateCount": parsed_dates.isna().sum(),
        "MinimumDate": parsed_dates.min(),
        "MaximumDate": parsed_dates.max()
    })

display(pd.DataFrame(date_review))


# =========================================================
# 10. CreatedOn and ModifiedOn Date Logic
# =========================================================

print("\n10. CREATED / MODIFIED DATE LOGIC")
print("=" * 80)

created_dates = pd.to_datetime(
    product_catalog_products_raw["CreatedOn"],
    errors="coerce"
)

modified_dates = pd.to_datetime(
    product_catalog_products_raw["ModifiedOn"],
    errors="coerce"
)

date_logic_issues = product_catalog_products_raw[
    modified_dates < created_dates
]

print(
    f"Records where ModifiedOn is earlier than CreatedOn: "
    f"{len(date_logic_issues)}"
)

if not date_logic_issues.empty:

    display(
        date_logic_issues[
            [
                "ProductCode",
                "CreatedOn",
                "ModifiedOn"
            ]
        ]
    )


# =========================================================
# 11. CategoryCode and SupplierCode Relationship Review
# =========================================================

print("\n11. RELATIONSHIP VALUE REVIEW")
print("=" * 80)

for column in ["CategoryCode", "SupplierCode"]:

    print(f"\n{column} frequency:")

    display(
        product_catalog_products_raw[column]
        .value_counts(dropna=False)
        .sort_index()
        .rename("RecordCount")
        .to_frame()
    )


# =========================================================
# 12. Source Datatype Review
# =========================================================

print("\n12. SOURCE DATATYPE REVIEW")
print("=" * 80)

datatype_review = pd.DataFrame({
    "Column": product_catalog_products_raw.columns,
    "SourceDtype": [
        str(product_catalog_products_raw[column].dtype)
        for column in product_catalog_products_raw.columns
    ]
})

display(datatype_review)


# =========================================================
# Independent Analyst Review Complete
# =========================================================

print("\n" + "=" * 80)
print("Independent Analyst Review completed.")
print("No data was modified.")
print("=" * 80)
```

## 3.1 Transformation Validation

Validate that every transformation approved in Section 1.8 was successfully applied to the clean Product dataset.

This validation checks:

- SKU whitespace removal
- ProductName whitespace removal
- CreatedOn datetime conversion
- ModifiedOn datetime conversion
- WeightKg outlier candidate retention

Only approved transformations are validated.

No additional data modifications are performed in this section.

```python
# =========================================================
# 3.1 Transformation Validation
# =========================================================

validation_results = []


# ---------------------------------------------------------
# 1. SKU Whitespace Validation
# ---------------------------------------------------------

sku_whitespace_remaining = (
    product_catalog_products_clean["SKU"]
    != product_catalog_products_clean["SKU"].str.strip()
).sum()

validation_results.append({
    "Validation": "SKU whitespace removal",
    "Expected": "0 records with leading/trailing whitespace",
    "Actual": f"{sku_whitespace_remaining} records remaining",
    "Passed": sku_whitespace_remaining == 0
})


# ---------------------------------------------------------
# 2. ProductName Whitespace Validation
# ---------------------------------------------------------

product_name_whitespace_remaining = (
    product_catalog_products_clean["ProductName"]
    != product_catalog_products_clean["ProductName"].str.strip()
).sum()

validation_results.append({
    "Validation": "ProductName whitespace removal",
    "Expected": "0 records with leading/trailing whitespace",
    "Actual": f"{product_name_whitespace_remaining} records remaining",
    "Passed": product_name_whitespace_remaining == 0
})


# ---------------------------------------------------------
# 3. CreatedOn Datetime Validation
# ---------------------------------------------------------

created_on_is_datetime = pd.api.types.is_datetime64_any_dtype(
    product_catalog_products_clean["CreatedOn"]
)

validation_results.append({
    "Validation": "CreatedOn datetime conversion",
    "Expected": "datetime64 datatype",
    "Actual": str(product_catalog_products_clean["CreatedOn"].dtype),
    "Passed": created_on_is_datetime
})


# ---------------------------------------------------------
# 4. ModifiedOn Datetime Validation
# ---------------------------------------------------------

modified_on_is_datetime = pd.api.types.is_datetime64_any_dtype(
    product_catalog_products_clean["ModifiedOn"]
)

validation_results.append({
    "Validation": "ModifiedOn datetime conversion",
    "Expected": "datetime64 datatype",
    "Actual": str(product_catalog_products_clean["ModifiedOn"].dtype),
    "Passed": modified_on_is_datetime
})


# ---------------------------------------------------------
# 5. WeightKg Retention Validation
# ---------------------------------------------------------

raw_p0010_weight = product_catalog_products_raw.loc[
    product_catalog_products_raw["ProductCode"] == "P0010",
    "WeightKg"
].iloc[0]

clean_p0010_weight = product_catalog_products_clean.loc[
    product_catalog_products_clean["ProductCode"] == "P0010",
    "WeightKg"
].iloc[0]

weight_retained = raw_p0010_weight == clean_p0010_weight

validation_results.append({
    "Validation": "WeightKg outlier candidate retention",
    "Expected": f"WeightKg remains {raw_p0010_weight}",
    "Actual": f"WeightKg is {clean_p0010_weight}",
    "Passed": weight_retained
})


# ---------------------------------------------------------
# Transformation Validation Results
# ---------------------------------------------------------

transformation_validation = pd.DataFrame(validation_results)

print("TRANSFORMATION VALIDATION RESULTS")
print("=" * 120)

display(transformation_validation)


# ---------------------------------------------------------
# Overall Result
# ---------------------------------------------------------

overall_validation_passed = transformation_validation["Passed"].all()

print(
    "\nOverall Transformation Validation Passed:",
    overall_validation_passed
)

if not overall_validation_passed:
    raise ValueError(
        "Transformation validation failed. Review the results "
        "before proceeding."
    )
```

## 3.2 Validate Data Preservation

Validate that the approved transformations did not unintentionally alter unaffected Product data.

The validation compares the raw and cleaned datasets while accounting for the approved transformations:

- SKU whitespace removal
- ProductName whitespace removal
- CreatedOn datatype conversion
- ModifiedOn datatype conversion

All other business and audit field values should remain unchanged.

```python
# =========================================================
# 3.2 Validate Data Preservation
# =========================================================

# Columns that should remain completely unchanged
unchanged_columns = [
    "ProductCode",
    "Description",
    "CategoryCode",
    "SupplierCode",
    "UnitCost",
    "ListPrice",
    "WeightKg",
    "WarrantyMonths",
    "ActiveFlag",
    "CreatedByUser",
    "ModifiedByUser"
]

all_columns_match = True

for column in unchanged_columns:

    matches = (
        product_catalog_products_raw[column]
        .astype(str)
        .reset_index(drop=True)
        ==
        product_catalog_products_clean[column]
        .astype(str)
        .reset_index(drop=True)
    ).all()

    print(f"{column}: {matches}")

    if not matches:
        all_columns_match = False


print(
    "\nAll unaffected columns preserved:",
    all_columns_match
)

if not all_columns_match:
    raise ValueError(
        "Data preservation validation failed. "
        "An unaffected column was unexpectedly changed."
    )
```

## 6.2.1 Create Product Catalog Table

The `commerce.product` table does not currently exist.

This section creates the Product Catalog PostgreSQL target table based

on the approved LJ Dev Commerce Product Data Dictionary.

Key design decisions:

- `product_id` is the Primary Key.

- All required Data Dictionary fields use `NOT NULL`.

- `category_id`, `supplier_id`, and `weight_kg` are optional and
  therefore allow `NULL`.

- `category_id` and `supplier_id` are Foreign Keys based on the
  approved Product database relationships.

- `is_active` uses the PostgreSQL `BOOLEAN` datatype.

- `created_date` and `updated_date` use the PostgreSQL `TIMESTAMP`
  datatype.

- `cost_price`, `selling_price`, and `weight_kg` use the PostgreSQL
  `DECIMAL` datatype.

```python
# =========================================================
# 6.2.1 Create Product Catalog Table
# =========================================================

create_product_catalog_table_query = """
CREATE TABLE commerce.product (

    product_id TEXT PRIMARY KEY,

    sku TEXT NOT NULL,
    product_name TEXT NOT NULL,
    product_description TEXT NOT NULL,

    category_id TEXT,
    supplier_id TEXT,

    cost_price DECIMAL NOT NULL,
    selling_price DECIMAL NOT NULL,

    weight_kg DECIMAL,

    warranty_months INTEGER NOT NULL,

    is_active BOOLEAN NOT NULL,

    created_date TIMESTAMP NOT NULL,
    created_by TEXT NOT NULL,

    updated_date TIMESTAMP NOT NULL,
    updated_by TEXT NOT NULL,

    FOREIGN KEY (category_id)
        REFERENCES commerce.category(category_id),

    FOREIGN KEY (supplier_id)
        REFERENCES commerce.supplier(supplier_id)

);
"""

try:
    cursor.execute(create_product_catalog_table_query)
    conn.commit()

    print("Product Catalog table created successfully.")
    print("Table: commerce.product")

except Exception as e:
    conn.rollback()

    print("Product Catalog table creation failed.")
    print("Error:", e)
```

Verify the applicable keys, relationships, and load dependencies before

```python
# =========================================================
# 6.4 Verify Keys, Relationships & Load Dependencies
# =========================================================

print("Product Catalog Keys, Relationships & Load Dependency Review")
print("=" * 80)

# ---------------------------------------------------------
# 1. Primary Key Verification
# ---------------------------------------------------------

cursor.execute("""
    SELECT
        tc.constraint_name,
        kcu.column_name
    FROM information_schema.table_constraints AS tc
    JOIN information_schema.key_column_usage AS kcu
        ON tc.constraint_name = kcu.constraint_name
        AND tc.table_schema = kcu.table_schema
    WHERE tc.table_schema = 'commerce'
      AND tc.table_name = 'product'
      AND tc.constraint_type = 'PRIMARY KEY';
""")

primary_keys = cursor.fetchall()

print("\n1. Primary Key")
print(primary_keys)

primary_key_valid = (
    len(primary_keys) == 1
    and primary_keys[0][1] == "product_id"
)

print("Primary Key validation passed:", primary_key_valid)


# ---------------------------------------------------------
# 2. Foreign Key Verification
# ---------------------------------------------------------

cursor.execute("""
    SELECT
        tc.constraint_name,
        kcu.column_name,
        ccu.table_schema AS referenced_schema,
        ccu.table_name AS referenced_table,
        ccu.column_name AS referenced_column
    FROM information_schema.table_constraints AS tc
    JOIN information_schema.key_column_usage AS kcu
        ON tc.constraint_name = kcu.constraint_name
        AND tc.table_schema = kcu.table_schema
    JOIN information_schema.constraint_column_usage AS ccu
        ON ccu.constraint_name = tc.constraint_name
        AND ccu.table_schema = tc.table_schema
    WHERE tc.table_schema = 'commerce'
      AND tc.table_name = 'product'
      AND tc.constraint_type = 'FOREIGN KEY'
    ORDER BY kcu.column_name;
""")

foreign_keys = cursor.fetchall()

print("\n2. Foreign Keys")
print(foreign_keys)

expected_foreign_key_columns = {
    "category_id",
    "supplier_id"
}

actual_foreign_key_columns = {
    row[1] for row in foreign_keys
}

foreign_keys_valid = (
    actual_foreign_key_columns
    == expected_foreign_key_columns
)

print("Expected Foreign Key columns:",
      expected_foreign_key_columns)

print("Actual Foreign Key columns:",
      actual_foreign_key_columns)

print("Foreign Key validation passed:",
      foreign_keys_valid)


# ---------------------------------------------------------
# 3. Load Dependency Review
# ---------------------------------------------------------

print("\n3. Load Dependency Review")

load_dependency_ready = (
    primary_key_valid
    and foreign_keys_valid
)

print("Primary Key ready:", primary_key_valid)

print("Foreign Key dependencies ready:",
      foreign_keys_valid)


# ---------------------------------------------------------
# 4. Overall Load Dependency Status
# ---------------------------------------------------------

print("\n4. Overall Load Dependency Status")
print("=" * 80)

print("Product Catalog is ready for data loading:",
      load_dependency_ready)
```

```python
# =========================================================
# 6.9 Product Catalog Database Integrity Validation
# =========================================================

integrity_query = """
SELECT
    COUNT(*) AS total_rows,

    COUNT(DISTINCT product_id) AS distinct_product_ids,

    COUNT(*) FILTER (
        WHERE product_id IS NULL
    ) AS null_product_ids,

    COUNT(*) FILTER (
        WHERE sku IS NULL
    ) AS null_skus,

    COUNT(*) FILTER (
        WHERE product_name IS NULL
    ) AS null_product_names,

    COUNT(*) FILTER (
        WHERE product_description IS NULL
    ) AS null_product_descriptions,

    COUNT(*) FILTER (
        WHERE cost_price IS NULL
    ) AS null_cost_prices,

    COUNT(*) FILTER (
        WHERE selling_price IS NULL
    ) AS null_selling_prices,

    COUNT(*) FILTER (
        WHERE warranty_months IS NULL
    ) AS null_warranty_months,

    COUNT(*) FILTER (
        WHERE is_active IS NULL
    ) AS null_active_flags,

    COUNT(*) FILTER (
        WHERE created_date IS NULL
    ) AS null_created_dates,

    COUNT(*) FILTER (
        WHERE created_by IS NULL
    ) AS null_created_by,

    COUNT(*) FILTER (
        WHERE updated_date IS NULL
    ) AS null_updated_dates,

    COUNT(*) FILTER (
        WHERE updated_by IS NULL
    ) AS null_updated_by,

    COUNT(*) FILTER (
        WHERE cost_price < 0
    ) AS negative_cost_prices,

    COUNT(*) FILTER (
        WHERE selling_price < 0
    ) AS negative_selling_prices,

    COUNT(*) FILTER (
        WHERE weight_kg < 0
    ) AS negative_weights,

    COUNT(*) FILTER (
        WHERE warranty_months < 0
    ) AS negative_warranty_months,

    COUNT(*) FILTER (
        WHERE updated_date < created_date
    ) AS invalid_audit_date_order

FROM commerce.product;
"""

try:
    cursor.execute(integrity_query)

    integrity_result = cursor.fetchone()

    print("Product Catalog Database Integrity Validation")
    print("=" * 80)

    print("\n1. Primary Key Integrity")
    print("Total rows:", integrity_result[0])
    print("Distinct product IDs:", integrity_result[1])
    print("NULL product IDs:", integrity_result[2])

    print("\n2. Required Field Integrity")
    print("NULL SKUs:", integrity_result[3])
    print("NULL product names:", integrity_result[4])
    print("NULL product descriptions:", integrity_result[5])
    print("NULL cost prices:", integrity_result[6])
    print("NULL selling prices:", integrity_result[7])
    print("NULL warranty months:", integrity_result[8])
    print("NULL active flags:", integrity_result[9])
    print("NULL created dates:", integrity_result[10])
    print("NULL created by:", integrity_result[11])
    print("NULL updated dates:", integrity_result[12])
    print("NULL updated by:", integrity_result[13])

    print("\n3. Product Value Integrity")
    print("Negative cost prices:", integrity_result[14])
    print("Negative selling prices:", integrity_result[15])
    print("Negative weights:", integrity_result[16])
    print("Negative warranty months:", integrity_result[17])

    print("\n4. Audit Date Integrity")
    print(
        "Updated date earlier than created date:",
        integrity_result[18]
    )

    # -----------------------------------------------------
    # Final Validation Result
    # -----------------------------------------------------

    integrity_passed = (
        integrity_result[0] == integrity_result[1]
        and integrity_result[2] == 0
        and integrity_result[3] == 0
        and integrity_result[4] == 0
        and integrity_result[5] == 0
        and integrity_result[6] == 0
        and integrity_result[7] == 0
        and integrity_result[8] == 0
        and integrity_result[9] == 0
        and integrity_result[10] == 0
        and integrity_result[11] == 0
        and integrity_result[12] == 0
        and integrity_result[13] == 0
        and integrity_result[14] == 0
        and integrity_result[15] == 0
        and integrity_result[16] == 0
        and integrity_result[17] == 0
        and integrity_result[18] == 0
    )

    print("\n5. Database Integrity Validation Summary")
    print("=" * 80)

    print(
        "Product Catalog database integrity passed:",
        integrity_passed
    )

except Exception as e:
    print("Product Catalog database integrity validation failed.")
    print("Error:", e)
```
