## 1.7 Independent Analyst Review

Perform a systematic independent analyst review of the untouched raw Shipment dataset.

This section is the notebook-based version of the Pre-ETL Inspection.

The review goes beyond the automated profiler and checks the Shipment dataset using independent analyst validation relevant to its business meaning and approved ETL design.

The review covers, where applicable:

- Dataset structure and grain

- Missing and blank values

- Exact duplicate rows

- Primary-key candidates and key uniqueness

- Parent order relationship and shipment-to-order consistency

- Whitespace in text fields

- Text consistency

- Shipment method and shipment status consistency

- Numeric validity and statistical outlier candidates

- Shipment amount and shipment status business-rule consistency

- Date parsing, datatype, validity, and Created/Modified date consistency

- Source-to-target datatype compatibility considerations

- Shipment relationship values

Automated profiler findings and independent analyst findings are kept conceptually separate.

The Shipment dataset is reviewed independently before any transformation.

No data is modified in this section.

```python
# =========================================================
# 1.7 Independent Analyst Review
# =========================================================

source_df = shipment_raw

print("1. DATASET STRUCTURE AND GRAIN")
print("=" * 80)
print(f"Rows: {source_df.shape[0]}")
print(f"Columns: {source_df.shape[1]}")
print("Working assumption: One row represents one shipment record.")

print("\n\n2. MISSING VALUES AND BLANK VALUES")
print("=" * 80)
missing_values = source_df.isna().sum()
blank_values = pd.Series({
    column: (source_df[column].astype("string").str.strip().eq("").sum()
             if source_df[column].dtype == "object" else 0)
    for column in source_df.columns
})
display(pd.DataFrame({"MissingValues": missing_values, "BlankValues": blank_values}))

print("\n\n3. EXACT DUPLICATE ROWS")
print("=" * 80)
exact_duplicate_count = source_df.duplicated().sum()
print("Shipment - Exact duplicate rows:", exact_duplicate_count)
if exact_duplicate_count > 0:
    display(source_df[source_df.duplicated(keep=False)])

print("\n\n4. PRIMARY KEY CANDIDATE AND IDENTIFIER UNIQUENESS")
print("=" * 80)
identifier_columns = ["ShipmentReference", "SourceOrderReference", "DispatchWarehouseCode"]
identifier_review = pd.DataFrame({
    "Column": identifier_columns,
    "NullCount": [source_df[c].isna().sum() for c in identifier_columns],
    "BlankCount": [source_df[c].astype("string").str.strip().eq("").sum() for c in identifier_columns],
    "DuplicateCount": [source_df[c].duplicated().sum() for c in identifier_columns],
    "UniqueValues": [source_df[c].nunique(dropna=True) for c in identifier_columns]
})
display(identifier_review)

print("\n\n5. SHIPMENT IDENTIFIER COLLISION REVIEW")
print("=" * 80)
duplicates = source_df[source_df["ShipmentReference"].duplicated(keep=False)]
print("Duplicate ShipmentReference records:", len(duplicates))
if not duplicates.empty:
    display(duplicates)

print("\n\n6. PARENT ORDER / WAREHOUSE RELATIONSHIP REVIEW")
print("=" * 80)
print("Distinct SourceOrderReference values:", source_df["SourceOrderReference"].nunique())
print("Distinct DispatchWarehouseCode values:", source_df["DispatchWarehouseCode"].nunique())
print("\nShipment records by source order:")
display(source_df.groupby("SourceOrderReference").size().reset_index(name="ShipmentCount").sort_values("ShipmentCount", ascending=False))
print("\nShipment records by warehouse:")
display(source_df.groupby("DispatchWarehouseCode").size().reset_index(name="ShipmentCount").sort_values("ShipmentCount", ascending=False))

print("\n\n7. LEADING AND TRAILING WHITESPACE REVIEW")
print("=" * 80)
text_columns = source_df.select_dtypes(include="object").columns
whitespace_findings=[]
for column in text_columns:
    mask=source_df[column].astype("string") != source_df[column].astype("string").str.strip()
    count=mask.sum()
    if count>0:
        whitespace_findings.append({"Column":column,"WhitespaceRecordCount":count})
        display(source_df.loc[mask,["ShipmentReference",column]])
if whitespace_findings:
    display(pd.DataFrame(whitespace_findings))
else:
    print("No leading or trailing whitespace found.")

print("\n\n8. TEXT / STATUS CONSISTENCY REVIEW")
print("=" * 80)
for column in ["Courier","TrackingStatus","DeliveryNotes","CreatedByUser","ModifiedByUser"]:
    print(f"\n{column} values:")
    display(source_df[column].value_counts(dropna=False).sort_index().rename("RecordCount").to_frame())

print("\n\n9. TRACKING NUMBER OPTIONALITY REVIEW")
print("=" * 80)
missing_tracking=source_df[source_df["TrackingNo"].isna()]
print("NULL TrackingNo records:",len(missing_tracking))
if not missing_tracking.empty:
    display(missing_tracking[["ShipmentReference","TrackingNo","DeliveryNotes"]])

print("\n\n10. DATE VALIDITY AND DATATYPE REVIEW")
print("=" * 80)
date_columns=["DispatchDate","DeliveredDate","CreatedOn","ModifiedOn"]
for column in date_columns:
    parsed=pd.to_datetime(source_df[column],errors="coerce")
    invalid=(parsed.isna() & source_df[column].notna()).sum()
    print(f"{column}: pandas dtype={source_df[column].dtype}, invalid populated dates={invalid}")

print("\n\n11. DATE RELATIONSHIP REVIEW")
print("=" * 80)
dispatch=pd.to_datetime(source_df["DispatchDate"],errors="coerce")
delivered=pd.to_datetime(source_df["DeliveredDate"],errors="coerce")
created=pd.to_datetime(source_df["CreatedOn"],errors="coerce")
modified=pd.to_datetime(source_df["ModifiedOn"],errors="coerce")
status=source_df["TrackingStatus"].astype("string").str.strip()
print("Delivered before DispatchDate:",(delivered.notna() & dispatch.notna() & (delivered<dispatch)).sum())
print("CreatedOn after DispatchDate:",(created.notna() & dispatch.notna() & (created>dispatch)).sum())
print("ModifiedOn before CreatedOn:",(modified.notna() & created.notna() & (modified<created)).sum())
print("Delivered shipments without DeliveredDate:",(status.eq("Delivered") & delivered.isna()).sum())
print("Non-delivered shipments with DeliveredDate",((~status.eq("Delivered")) & delivered.notna()).sum())

print("\n\n12. STATUS / DELIVERY-DATE BUSINESS RULE REVIEW")
print("=" * 80)
print("Observed trimmed shipment statuses:", sorted(status.dropna().unique().tolist()))
display(source_df.loc[source_df["ShipmentReference"].eq("SH013"),["ShipmentReference","TrackingStatus","DeliveredDate","DeliveryNotes"]])

print("\n\n13. SOURCE-TO-TARGET DATATYPE PREPARATION REVIEW")
print("=" * 80)
datatype_review=pd.DataFrame([
    ["ShipmentReference",str(source_df["ShipmentReference"].dtype),"Text","Retain"],
    ["SourceOrderReference",str(source_df["SourceOrderReference"].dtype),"Text","Retain"],
    ["DispatchWarehouseCode",str(source_df["DispatchWarehouseCode"].dtype),"Text","Retain"],
    ["DispatchDate",str(source_df["DispatchDate"].dtype),"Datetime","Convert"],
    ["Courier",str(source_df["Courier"].dtype),"Text","Retain"],
    ["TrackingNo",str(source_df["TrackingNo"].dtype),"Text","Preserve NULL"],
    ["TrackingStatus",str(source_df["TrackingStatus"].dtype),"Text","Trim + standardize"],
    ["DeliveredDate",str(source_df["DeliveredDate"].dtype),"Datetime","Convert; preserve NULL"],
    ["DeliveryNotes",str(source_df["DeliveryNotes"].dtype),"Text","Retain"],
    ["CreatedOn",str(source_df["CreatedOn"].dtype),"Datetime","Convert"],
    ["CreatedByUser",str(source_df["CreatedByUser"].dtype),"Text","Retain"],
    ["ModifiedOn",str(source_df["ModifiedOn"].dtype),"Datetime","Convert"],
    ["ModifiedByUser",str(source_df["ModifiedByUser"].dtype),"Text","Retain"]
],columns=["SourceColumn","CurrentPandasDtype","TargetType","Preparation"])
display(datatype_review)

print("\nIndependent analyst review completed.")
```

```python
# =========================================================
# 2.1 Create Clean Working Dataset
# =========================================================

shipment_clean=shipment_raw.copy()

print("Clean working dataset created successfully.")
print("Shipment raw rows:",shipment_raw.shape[0])
print("Shipment clean rows:",shipment_clean.shape[0])
print("Raw and clean datasets are separate objects:",shipment_raw is not shipment_clean)
```

## 2.2 Apply Approved Transformations

Apply only the transformations approved in Section 1.8 to the clean working Shipment dataset.

Approved transformations:

- Trim leading/trailing whitespace from `TrackingStatus`.
- Standardize SH013 `TrackingStatus` to `In Transit`.
- Convert `DispatchDate`, `DeliveredDate`, `CreatedOn`, and `ModifiedOn` to datetime.
- Preserve legitimate NULL `TrackingNo` values.
- Preserve SH013 `DeliveredDate` as NULL.

No other business values are modified.


```python
# =========================================================
# 2.2 Apply Approved Transformations
# =========================================================

shipment_clean["TrackingStatus"] = shipment_clean["TrackingStatus"].astype("string").str.strip()
shipment_clean["TrackingStatus"] = shipment_clean["TrackingStatus"].replace({"in transit":"In Transit"})

for column in ["DispatchDate","DeliveredDate","CreatedOn","ModifiedOn"]:
    shipment_clean[column] = pd.to_datetime(shipment_clean[column], errors="raise")

# TrackingNo NULL values are intentionally preserved.

print("Approved Shipment transformations applied.")
```

## 3.1 Transformation Validation

Validate every approved Shipment transformation from Section 1.8.

Checks include status standardization, datetime conversion, preservation of legitimate TrackingNo NULLs, preservation of SH013 delivery-date NULL, and approved shipment business rules.


```python
# =========================================================
# 3.1 Transformation Validation
# =========================================================

allowed_statuses={"Delivered","In Transit"}
status_valid=set(shipment_clean["TrackingStatus"].dropna().unique()) <= allowed_statuses
sh013=shipment_clean.loc[shipment_clean["ShipmentReference"].eq("SH013")]
sh013_status_valid=(len(sh013)==1 and sh013.iloc[0]["TrackingStatus"]=="In Transit")
sh013_delivery_null=(len(sh013)==1 and pd.isna(sh013.iloc[0]["DeliveredDate"]))
datetime_valid=all(pd.api.types.is_datetime64_any_dtype(shipment_clean[c]) for c in ["DispatchDate","DeliveredDate","CreatedOn","ModifiedOn"])
tracking_null_ids=set(shipment_clean.loc[shipment_clean["TrackingNo"].isna(),"ShipmentReference"])
tracking_nulls_valid=(tracking_null_ids=={"SH007","SH012"})

print("Shipment Transformation Validation")
print("Status values valid:",status_valid)
print("SH013 standardized to In Transit:",sh013_status_valid)
print("Datetime fields valid:",datetime_valid)
print("SH013 DeliveredDate remains NULL:",sh013_delivery_null)
print("Tracking NULLs preserved:",tracking_nulls_valid)
transformation_validation_passed=all([status_valid,sh013_status_valid,datetime_valid,sh013_delivery_null,tracking_nulls_valid])
print("Overall transformation validation passed:",transformation_validation_passed)
```

## 3.2 Validate Data Preservation

Validate that the approved Shipment transformations did not unintentionally alter unaffected source data.

The raw and clean datasets are compared while accounting for the approved status and datatype transformations.


```python
# =========================================================
# 3.2 Validate Data Preservation
# =========================================================

preserved_columns=["ShipmentReference","SourceOrderReference","DispatchWarehouseCode","Courier","TrackingNo","DeliveryNotes","CreatedByUser","ModifiedByUser"]
row_count_preserved=len(shipment_clean)==len(shipment_raw)
column_structure_preserved=shipment_clean.columns.tolist()==shipment_raw.columns.tolist()
preservation_results={c: shipment_clean[c].astype("string").equals(shipment_raw[c].astype("string")) for c in preserved_columns}
preserved_values_valid=all(preservation_results.values())
shipment_ids_preserved=shipment_clean["ShipmentReference"].tolist()==shipment_raw["ShipmentReference"].tolist()
data_preservation_passed=all([row_count_preserved,column_structure_preserved,preserved_values_valid,shipment_ids_preserved])

display(pd.Series(preservation_results,name="Preserved"))
print("Row count preserved:",row_count_preserved)
print("Column structure preserved:",column_structure_preserved)
print("Shipment IDs preserved:",shipment_ids_preserved)
print("Unchanged values preserved:",preserved_values_valid)
print("Overall data preservation validation passed:",data_preservation_passed)
```

## 6.2 Prepare / Create Target Table

Check whether the approved PostgreSQL target table already exists before creating or modifying anything.

Target:

Schema: `commerce`

Table: `shipment`

The table must not be recreated blindly.


```python
# =========================================================
# 6.2 Prepare / Create Target Table
# =========================================================

target_schema = "commerce"
target_table = "shipment"

# Check whether the target table already exists
cursor.execute("""
    SELECT EXISTS (
        SELECT 1
        FROM information_schema.tables
        WHERE table_schema = %s
          AND table_name = %s
    );
""", (target_schema, target_table))

table_exists = cursor.fetchone()[0]

print("Shipment Target Table Check")
print("=" * 80)

print("Schema:", target_schema)
print("Table:", target_table)
print("Table exists:", table_exists)

if table_exists:
    print("\nTarget table already exists.")
    print("The table will not be recreated.")
else:
    print("\nTarget table does not exist.")
    print("The approved database design must be used before creating it.")
```

## 6.2.1 Create Shipment Table

Create `commerce.shipment` only when the target table does not already exist.

The table follows the current Shipment Data Dictionary. `delivery_date` and `tracking_number` are nullable; required fields use `NOT NULL`; `shipment_id` is the Primary Key; `sales_order_id` and `warehouse_id` are Foreign Keys.


```python
# =========================================================
# 6.2.1 Create Shipment Table
# =========================================================

if not table_exists:
    create_shipment_table_query="""
    CREATE TABLE commerce.shipment (
        shipment_id TEXT PRIMARY KEY,
        sales_order_id TEXT NOT NULL,
        warehouse_id TEXT NOT NULL,
        shipment_date TIMESTAMP NOT NULL,
        courier_name TEXT NOT NULL,
        tracking_number TEXT,
        shipment_status TEXT NOT NULL,
        delivery_date TIMESTAMP,
        notes TEXT,
        created_date TIMESTAMP NOT NULL,
        created_by TEXT NOT NULL,
        updated_date TIMESTAMP NOT NULL,
        updated_by TEXT NOT NULL,
        FOREIGN KEY (sales_order_id) REFERENCES commerce.sales_order(sales_order_id),
        FOREIGN KEY (warehouse_id) REFERENCES commerce.warehouse(warehouse_id)
    );
    """
    try:
        cursor.execute(create_shipment_table_query)
        conn.commit()
        print("Shipment table created successfully.")
    except Exception as e:
        conn.rollback()
        print("Shipment table creation failed.")
        print("Error:",e)
else:
    print("Shipment table already exists. Creation skipped.")
```

Relationships:
- `sales_order_id` → `commerce.sales_order.sales_order_id`
- `warehouse_id` → `commerce.warehouse.warehouse_id`


```python
# =========================================================
# 6.4 Verify Keys, Relationships & Load Dependencies
# =========================================================

print("Shipment Keys, Relationships & Load Dependency Review")
print("="*80)

cursor.execute("""
SELECT tc.constraint_name,kcu.column_name
FROM information_schema.table_constraints AS tc
JOIN information_schema.key_column_usage AS kcu
 ON tc.constraint_name=kcu.constraint_name AND tc.table_schema=kcu.table_schema
WHERE tc.table_schema='commerce' AND tc.table_name='shipment' AND tc.constraint_type='PRIMARY KEY';
""")
primary_keys=cursor.fetchall()
primary_key_valid=(len(primary_keys)==1 and primary_keys[0][1]=="shipment_id")
print("Primary Key:",primary_keys)
print("Primary Key validation passed:",primary_key_valid)

cursor.execute("""
SELECT tc.constraint_name,kcu.column_name,ccu.table_schema,ccu.table_name,ccu.column_name
FROM information_schema.table_constraints AS tc
JOIN information_schema.key_column_usage AS kcu
 ON tc.constraint_name=kcu.constraint_name AND tc.table_schema=kcu.table_schema
JOIN information_schema.constraint_column_usage AS ccu
 ON ccu.constraint_name=tc.constraint_name AND ccu.table_schema=tc.table_schema
WHERE tc.table_schema='commerce' AND tc.table_name='shipment' AND tc.constraint_type='FOREIGN KEY'
ORDER BY kcu.column_name;
""")
foreign_keys=cursor.fetchall()
expected_foreign_keys={("sales_order_id","sales_order","sales_order_id"),("warehouse_id","warehouse","warehouse_id")}
actual_foreign_keys={(r[1],r[3],r[4]) for r in foreign_keys}
foreign_keys_valid=actual_foreign_keys==expected_foreign_keys
print("Foreign Keys:",foreign_keys)
print("Foreign Key validation passed:",foreign_keys_valid)

sales_order_ids=shipment_db_ready["sales_order_id"].dropna().unique().tolist()
warehouse_ids=shipment_db_ready["warehouse_id"].dropna().unique().tolist()

placeholders=",".join(["%s"]*len(sales_order_ids))
cursor.execute(f"SELECT sales_order_id FROM commerce.sales_order WHERE sales_order_id IN ({placeholders});",sales_order_ids)
existing_sales_order_ids={r[0] for r in cursor.fetchall()}
missing_sales_order_ids=set(sales_order_ids)-existing_sales_order_ids
sales_order_dependency_valid=len(missing_sales_order_ids)==0

placeholders=",".join(["%s"]*len(warehouse_ids))
cursor.execute(f"SELECT warehouse_id FROM commerce.warehouse WHERE warehouse_id IN ({placeholders});",warehouse_ids)
existing_warehouse_ids={r[0] for r in cursor.fetchall()}
missing_warehouse_ids=set(warehouse_ids)-existing_warehouse_ids
warehouse_dependency_valid=len(missing_warehouse_ids)==0

print("Missing Sales Order IDs:",missing_sales_order_ids)
print("Sales Order dependency ready:",sales_order_dependency_valid)
print("Missing Warehouse IDs:",missing_warehouse_ids)
print("Warehouse dependency ready:",warehouse_dependency_valid)

load_dependency_ready=all([primary_key_valid,foreign_keys_valid,sales_order_dependency_valid,warehouse_dependency_valid])
print("Shipment is ready for data loading:",load_dependency_ready)
```

## 6.5 Insert Shipment Records

Load the validated Shipment Database-Ready dataset into the verified `commerce.shipment` PostgreSQL table.

Before inserting, confirm the target table is empty, confirm the database-ready row count, insert only validated records, use the approved target column list, and commit only after successful insertion.


```python
# =========================================================
# 6.5 Insert Shipment Records
# =========================================================

print("Shipment Record Loading")
print("="*80)

cursor.execute("SELECT COUNT(*) FROM commerce.shipment;")
existing_record_count=cursor.fetchone()[0]
target_table_empty=existing_record_count==0
db_ready_row_count=len(shipment_db_ready)

print("Existing records:",existing_record_count)
print("Target table is empty:",target_table_empty)
print("Database-ready records:",db_ready_row_count)

insert_shipment_query="""
INSERT INTO commerce.shipment (
    shipment_id,sales_order_id,warehouse_id,shipment_date,courier_name,
    tracking_number,shipment_status,delivery_date,notes,created_date,
    created_by,updated_date,updated_by
)
VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s);
"""

try:
    if not target_table_empty:
        raise ValueError("Target table is not empty. Insertion stopped to prevent accidental duplicate loading.")

    records_to_insert=[]
    
    for row in shipment_db_ready.itertuples(index=False,name=None):
        cleaned_row=tuple(
            None if pd.isna(value) else value
            for value in row
        )
        records_to_insert.append(cleaned_row)

    if len(records_to_insert)!=db_ready_row_count:
        raise ValueError("Database-ready record count changed before insertion.")

    cursor.executemany(insert_shipment_query,records_to_insert)
    conn.commit()

    print("Shipment records inserted:",len(records_to_insert))
    print("Transaction committed successfully.")

except Exception as e:
    conn.rollback()
    print("Shipment record insertion failed.")
    print("Transaction rolled back.")
    print("Error:",e)
```

## 6.9 Shipment Database Integrity Validation

Perform final integrity checks on `commerce.shipment`, including primary-key uniqueness, required fields, foreign-key relationships, approved status values, Delivered → delivery-date requirement, In Transit → nullable delivery-date behavior, and audit/date consistency.


```python
# =========================================================
# 6.9 Shipment Database Integrity Validation
# =========================================================

integrity_query="""
SELECT
 COUNT(*) AS total_rows,
 COUNT(DISTINCT shipment_id) AS distinct_shipment_ids,
 COUNT(*) FILTER (WHERE shipment_id IS NULL) AS null_shipment_ids,
 COUNT(*) FILTER (WHERE sales_order_id IS NULL) AS null_sales_order_ids,
 COUNT(*) FILTER (WHERE warehouse_id IS NULL) AS null_warehouse_ids,
 COUNT(*) FILTER (WHERE shipment_date IS NULL) AS null_shipment_dates,
 COUNT(*) FILTER (WHERE courier_name IS NULL) AS null_courier_names,
 COUNT(*) FILTER (WHERE shipment_status IS NULL) AS null_shipment_statuses,
 COUNT(*) FILTER (WHERE created_date IS NULL) AS null_created_dates,
 COUNT(*) FILTER (WHERE created_by IS NULL) AS null_created_by,
 COUNT(*) FILTER (WHERE updated_date IS NULL) AS null_updated_dates,
 COUNT(*) FILTER (WHERE updated_by IS NULL) AS null_updated_by,
 COUNT(*) FILTER (WHERE shipment_status NOT IN ('Delivered','In Transit')) AS invalid_statuses,
 COUNT(*) FILTER (WHERE shipment_status='Delivered' AND delivery_date IS NULL) AS delivered_without_delivery_date,
 COUNT(*) FILTER (WHERE shipment_status='In Transit' AND delivery_date IS NOT NULL) AS in_transit_with_delivery_date,
 COUNT(*) FILTER (WHERE updated_date < created_date) AS invalid_audit_date_order,
 COUNT(*) FILTER (WHERE delivery_date IS NOT NULL AND delivery_date < shipment_date) AS delivery_before_shipment
FROM commerce.shipment;
"""

try:
    cursor.execute(integrity_query)
    r=cursor.fetchone()
    print("Shipment Database Integrity Validation")
    print("="*80)
    labels=["Total rows","Distinct shipment IDs","NULL shipment IDs","NULL sales order IDs","NULL warehouse IDs","NULL shipment dates","NULL courier names","NULL shipment statuses","NULL created dates","NULL created by","NULL updated dates","NULL updated by","Invalid statuses","Delivered without delivery date","In Transit with delivery date","Updated before created","Delivery before shipment"]
    for label,value in zip(labels,r): print(f"{label}: {value}")
    integrity_passed=(r[0]==r[1] and all(v==0 for v in r[2:]))
    print("\nShipment database integrity passed:",integrity_passed)
except Exception as e:
    print("Shipment database integrity validation failed.")
    print("Error:",e)
```
